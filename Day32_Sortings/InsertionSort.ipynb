# Insertion Sort
How it works:
1. Start with the second element (index 1) as the first element is considered sorted
2. Compare the current element with the previous elements in the sorted portion
3. Shift all larger elements one position to the right
4. Insert the current element in its correct position
5. Repeat for all remaining elements until the entire array is sorted

Time complexity: O(n²) worst case, O(n) best case (already sorted)
Space complexity: O(1) - sorts in place


```python
def insertionSort(arr):
    n = len(arr)

    if n <= 1:
        return arr
    
    print(f"Starting insertion sort with array: {arr}")
    print("=" * 50)
    
    for i in range(1,n):
        key = arr[i]
        print(f"\nStep {i}: Processing element at index {i}")
        print(f"Key to insert: {key}")
        print(f"Current array: {arr}")
        
        j = i-1
        print(f"Starting comparison from index {j}")
        
        while j >= 0 and key < arr[j]:
            print(f"  Comparing {key} < {arr[j]} at index {j}: True")
            print(f"  Shifting {arr[j]} from index {j} to index {j+1}")
            arr[j+1] = arr[j]
            print(f"  Array after shift: {arr}")
            j -= 1
            if j >= 0:
                print(f"  Moving to compare with index {j}")
        
        if j >= 0:
            print(f"  Comparing {key} < {arr[j]} at index {j}: False")
        
        arr[j+1] = key
        print(f"  Inserting {key} at index {j+1}")
        print(f"  Array after insertion: {arr}")
        print("-" * 30)

    print(f"\nFinal sorted array: {arr}")
    return arr

arr= [12, 11, 13, 5, -1]
print(insertionSort(arr))
```

```python
def insertion_sort(arr):
    n = len(arr)

    for i in range(1,n):
        j = i-1
        key = arr[i]
        while j >= 0 and key < arr[j]:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key
    return arr

print(insertion_sort([12, 11, 13, 5, -1]))
```
