70. Climbing Stairs

You are climbing a staircase. It takes n steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

 

Example 1:

Input: n = 2
Output: 2
Explanation: There are two ways to climb to the top.
1. 1 step + 1 step
2. 2 steps

Example 2:

Input: n = 3
Output: 3
Explanation: There are three ways to climb to the top.
1. 1 step + 1 step + 1 step
2. 1 step + 2 steps
3. 2 steps + 1 step
 

Constraints:

1 <= n <= 45

```python
def wyas(n):
    if n <= 1:
        return 1
    dp = [0]*(n+1)
    dp[0] = 1
    dp[1] = 1

    for i in range(2,n+1):
        dp[i] = dp[i-1]+dp[i-2]
    
    return dp[n]

print(wyas(6))
```

```python
# This function calculates the number of ways to climb n stairs using dynamic programming with memoization
# It's solving the classic "climbing stairs" problem where you can take 1 or 2 steps at a time
def ways2(n):
    # Dictionary to store previously calculated results to avoid redundant calculations
    memo = {}
    # Helper function that does the actual recursive calculation
    def helper(n):

        # Base case: if n is 0 or 1, there's only 1 way (stay put or take 1 step)
        if n <=1:
            return 1
        # If we've already calculated this value, return it from memo (memoization)
        if n in memo:
            return memo[n]
        
        # Recursive case: ways to reach step n = ways to reach (n-1) + ways to reach (n-2)
        # This is because from step (n-1) you can take 1 step, from step (n-2) you can take 2 steps
        memo[n] = helper(n-1) + helper(n-2)
        return memo[n]
    
    # Call the helper function with the input n
    return helper(n)

# Test the function with n=5, which should return 8 ways to climb 5 stairs
print(ways2(5))
```

```python
def ways3(n):
    if n <= 1:
        return 1
    
    prev1 = 1
    prev2 = 1

    for i in range(2, n+1):
        temp = prev1+prev2
        prev1 = prev2
        prev2 = temp
    
    return prev2

print(ways3(5))
```
