98. Validate Binary Search Tree

Given the root of a binary tree, determine if it is a valid binary search tree (BST).

A valid BST is defined as follows:

The left subtree of a node contains only nodes with keys strictly less than the node's key.
The right subtree of a node contains only nodes with keys strictly greater than the node's key.

Both the left and right subtrees must also be binary search trees.
 

Example 1:

Input: root = [2,1,3]
Output: true

Example 2:

Input: root = [5,1,4,null,null,3,6]
Output: false
Explanation: The root node's value is 5 but its right child's value is 4.
 
Constraints:

The number of nodes in the tree is in the range [1, 104].
-231 <= Node.val <= 231 - 1

# problem statement

here we go through the each element recursively,

- same for the right.
- now the first root value check completles then it also check for the following childern node till end 



```python
from typing import Optional

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def isValidBST(self, root: Optional[TreeNode]) -> bool:

        def isValid(Bst, min_value = float("-inf"), max_value = float("inf")):
            if not Bst:
                return True
            if not (min_value < Bst.val < max_value):
                return False
            
            return isValid(Bst.left, min_value, Bst.val) and isValid(Bst.right, Bst.val, max_value)

        return isValid(root)
# Test cases for the isValidBST function
def test_isValidBST():
    solution = Solution()
    
    # Test case 1: Empty tree
    assert solution.isValidBST(None) is True
    
    # Test case 2: Single node tree
    root1 = TreeNode(5)
    assert solution.isValidBST(root1) == True
    
    # Test case 3: Valid BST
    #     5
    #    / \
    #   3   7
    #  / \   \
    # 2   4   8
    root2 = TreeNode(5)
    root2.left = TreeNode(3)
    root2.right = TreeNode(7)
    root2.left.left = TreeNode(2)
    root2.left.right = TreeNode(4)
    root2.right.right = TreeNode(8)
    assert solution.isValidBST(root2) == True
    
    # Test case 4: Invalid BST - left child greater than parent
    #     5
    #    / \
    #   6   7
    root3 = TreeNode(5)
    root3.left = TreeNode(6)
    root3.right = TreeNode(7)
    assert solution.isValidBST(root3) == False
    
    # Test case 5: Invalid BST - right child less than parent
    #     5
    #    / \
    #   3   4
    root4 = TreeNode(5)
    root4.left = TreeNode(3)
    root4.right = TreeNode(4)
    assert solution.isValidBST(root4) == False
    
    # Test case 6: Left subtree only
    #     5
    #    /
    #   3
    #  /
    # 1
    root5 = TreeNode(5)
    root5.left = TreeNode(3)
    root5.left.left = TreeNode(1)
    assert solution.isValidBST(root5) == True
    
    # Test case 7: Right subtree only
    #   5
    #    \
    #     7
    #      \
    #       9
    root6 = TreeNode(5)
    root6.right = TreeNode(7)
    root6.right.right = TreeNode(9)
    assert solution.isValidBST(root6) == True

if __name__ == "__main__":
    test_isValidBST()
    print("All test cases passed!")
```

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        return self._traverse(root, float('-inf'), float('inf'))

    def _traverse(self, root: Optional[TreeNode], minimum: float, maximum: float) -> bool:
        if not root:
            return True
        else:
            left = self._traverse(root.left, minimum, root.val)
            right = self._traverse(root.right, root.val, maximum)
            return left and right and (minimum < root.val) and (root.val < maximum)
```
