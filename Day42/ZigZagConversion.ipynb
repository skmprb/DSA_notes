P   A   H   N
A P L S I I G
Y   I   R

And then read line by line: "PAHNAPLSIIGYIR"

Write the code that will take a string and make this conversion given a number of rows:

string convert(string s, int numRows);
 

Example 1:

Input: s = "PAYPALISHIRING", numRows = 3
Output: "PAHNAPLSIIGYIR"
Example 2:

Input: s = "PAYPALISHIRING", numRows = 4

Explanation:

A   L S  I G
Y A   H R
P     I

Example 3:

Input: s = "A", numRows = 1

Constraints:

1 <= s.length <= 1000
s consists of English letters (lower-case and upper-case), ',' and '.'.

```python
def convert(s,numRows):
    row = [""] * numRows
    current_row = 0
    going_down = False

    if len(s) <= 1 or numRows <= 1:
        return s
    
    if len(s) == numRows:
        return s

    for i in range(len(s)):
        row[current_row] += s[i]

        if current_row == 0:
            going_down = True
        if current_row == numRows-1:
            going_down = False
        
        if going_down == True:
            current_row += 1
        if going_down == False:
            current_row -= 1
    return "".join(row)


print(convert("PAYPALISHIRING",3))
print(convert("PAYPALISHIRING",4))
print(convert("PAYPALISHIRING",1))
print(convert("PAYPALISHIRING",14))
print(convert("P",1))
print(convert("",1))
```

# 2D Array Approach

This approach creates a 2D grid and fills it in zigzag pattern, then reads row by row.

```python
def convert_2d(s, numRows):
    if len(s) <= 1 or numRows <= 1:
        return s
    
    # Calculate number of columns needed
    # Each cycle: numRows down + (numRows-2) diagonal up
    cycle_len = 2 * numRows - 2
    num_cycles = (len(s) + cycle_len - 1) // cycle_len
    num_cols = num_cycles * (numRows - 1) if numRows > 1 else len(s)
    
    # Create 2D array filled with empty strings
    grid = [['' for _ in range(num_cols)] for _ in range(numRows)]
    
    row, col = 0, 0
    going_down = True
    
    for char in s:
        grid[row][col] = char
        
        if going_down:
            if row == numRows - 1:
                # Reached bottom, start going diagonal up
                going_down = False
                row -= 1
                col += 1
            else:
                row += 1
        else:
            if row == 0:
                # Reached top, start going down
                going_down = True
                row += 1
            else:
                row -= 1
                col += 1
    
    # Read row by row
    result = ''
    for r in range(numRows):
        for c in range(num_cols):
            if grid[r][c] != '':
                result += grid[r][c]
    
    return result

# Test cases
print("2D Array Approach:")
print(convert_2d("PAYPALISHIRING", 3))  # PAHNAPLSIIGYIR
print(convert_2d("PAYPALISHIRING", 4))  # PINALSIGYAHRPI
print(convert_2d("A", 1))  # A
print(convert_2d("PAYPALISHIRING", 1))  # PAYPALISHIRING
```

```python
class Solution:
    def convert(self, s: str, numRows: int) -> str:
        if numRows ==1:
            return s
        
        rows = [""] * numRows
        cur_row = 0
        direction=1
        for c in s:
            rows[cur_row]+= c

            if cur_row==0:
                direction = 1
            if cur_row==numRows-1:
                direction = -1

            cur_row+=direction

        return "".join(rows)
        
```
