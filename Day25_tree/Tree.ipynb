```python
## Main tree blueprint

class Node:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def solve(node):
    if not node:
        return []
    
    left = solve(node.left)
    right = solve(node.right)

    return (left, right , node)

node = Node(5)
node.left = Node(5)
node.right = Node(7)

print(solve(node))
```

```python
# height of the tree

class Node:
    def __init__ (self, val = 0, left = None,right= None):
        self.val = val
        self. left = left
        self.right = right

def height(node):
    if not node:
        return 0
    
    left = height(node.left)
    right = height(node.right)

    return max(left,right)+1

node = Node(5, Node(5), Node(7))
node.right = Node(6)
node.right.right = Node(8)
node.left = Node(9)
print(height(node))
```

```python
# Balanced and height

# Tree Balanced means
    # 1. left subtree height and right subtree height diff should be <= 1
    # 2. left subtree should be balanced
    # 3. right subtree should be balanced

# TreeNode class represents a node in a binary tree with a value and left/right child pointers
class TreeNode:
    def __init__ (self, val=0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

# Function to check if a binary tree is balanced and return its height
def idBalanced(root):
    # Helper function that returns a tuple: (height, is_balanced)
    def helper(node):
        # Base case: empty node has height 0 and is balanced
        if not node:
            return (0,True)

        # Recursively get height and balance status of left and right subtrees
        left_height, left_balanced = helper(node.left)
        right_height, right_balanced = helper(node.right)

        # Current node's height is 1 + maximum height of its subtrees
        height = max(left_height,right_height) + 1

        # Node is balanced if both subtrees are balanced AND height difference is <= 1
        isbalanced = (left_balanced and right_balanced and abs(left_height - right_height) <= 1)

        return (height, isbalanced)
    
    # Call helper function starting from root
    return helper(root)


# Create a sample binary tree for testing
root = TreeNode(5)
root.left = TreeNode(8)
root.right = TreeNode (10)
root.left.left = TreeNode(3)
root.left.right = TreeNode(4)
#root.right.right = TreeNode(6)
#root.right.left = TreeNode(12)
root.left.left.left = TreeNode(13)

# Print the result: (height, is_balanced) tuple for the tree
print(idBalanced(root))
    
```
