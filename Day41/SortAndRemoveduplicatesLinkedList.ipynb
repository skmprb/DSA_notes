83. Remove Duplicates from Sorted List

Given the head of a sorted linked list, delete all duplicates such that each element appears only once. Return the linked list sorted as well.

 

Example 1:


Input: head = [1,1,2]
Output: [1,2]
Example 2:


Input: head = [1,1,2,3,3]
Output: [1,2,3]
 

Constraints:

The number of nodes in the list is in the range [0, 300].
-100 <= Node.val <= 100
The list is guaranteed to be sorted in ascending order.

# The Problem solving

Here the given Linked list is sorted and here we have to remove the duplicates.

Duplicaton:
- we will do a while loop and check the next value, if the current and next value is same we will skip it by next.
- here we have run a while loopinside it the while for the consicutive same numbers.
- edge cases are no node, or only one node. 



```python
from typing import Optional, List

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def deleteDuplicates(self, head: Optional[ListNode]) -> Optional[ListNode]:
        if not head:
            return None
        
        current = head
        while current and current.next:
            if current.val == current.next.val:
                current.next = current.next.next
            else:
                current = current.next
        return head


# Testing the code with some linked list nodes
def print_list(head):
    """Helper function to print the linked list"""
    result = []
    current = head
    while current:
        result.append(current.val)
        current = current.next
    return result
        
# Test case 1: [1, 1, 2]
node1 = ListNode(1)
node2 = ListNode(1)
node3 = ListNode(2)
node1.next = node2
node2.next = node3

solution = Solution()
result1 = solution.deleteDuplicates(node1)
print("Test 1 - Input: [1, 1, 2], Output:", print_list(result1))

# Test case 2: [1, 1, 2, 3, 3]
node4 = ListNode(1)
node5 = ListNode(1)
node6 = ListNode(2)
node7 = ListNode(3)
node8 = ListNode(3)
node4.next = node5
node5.next = node6
node6.next = node7
node7.next = node8

result2 = solution.deleteDuplicates(node4)
print("Test 2 - Input: [1, 1, 2, 3, 3], Output:", print_list(result2))
# Test case 3: Empty list
result3 = solution.deleteDuplicates(None)
print("Test 3 - Input: [], Output:", result3)

# Test case 4: Single node [1]
node9 = ListNode(1)
result4 = solution.deleteDuplicates(node9)
print("Test 4 - Input: [1], Output:", print_list(result4))

# Test case 5: [2, 2, 2, 2, 3, 4, 4, 4, 4, 4] - 4 consecutive duplicates of 2 and 5 consecutive duplicates of 4
node10 = ListNode(2)
node11 = ListNode(2)
node12 = ListNode(2)
node13 = ListNode(2)
node14 = ListNode(3)
node15 = ListNode(4)
node16 = ListNode(4)
node17 = ListNode(4)
node18 = ListNode(4)
node19 = ListNode(4)
node10.next = node11
node11.next = node12
node12.next = node13
node13.next = node14
node14.next = node15
node15.next = node16
node16.next = node17
node17.next = node18
node18.next = node19

result5 = solution.deleteDuplicates(node10)
print("Test 5 - Input: [2, 2, 2, 2, 3, 4, 4, 4, 4, 4], Output:", print_list(result5))

# Test case 6: [1, 1, 1, 1, 1, 1] - 6 consecutive duplicates
node20 = ListNode(1)
node21 = ListNode(1)
node22 = ListNode(1)
node23 = ListNode(1)
node24 = ListNode(1)
node25 = ListNode(1)
node20.next = node21
node21.next = node22
node22.next = node23
node23.next = node24
node24.next = node25

result6 = solution.deleteDuplicates(node20)
print("Test 6 - Input: [1, 1, 1, 1, 1, 1], Output:", print_list(result6))


```
