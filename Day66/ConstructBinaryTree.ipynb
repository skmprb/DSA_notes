105. Construct Binary Tree from Preorder and Inorder Traversal

Given two integer arrays preorder and inorder where preorder is the preorder traversal of a binary tree and inorder is the inorder traversal of the same tree, construct and return the binary tree.

Example 1:

Input: preorder = [3,9,20,15,7], inorder = [9,3,15,20,7]
Output: [3,9,20,null,null,15,7]

Example 2:

Input: preorder = [-1], inorder = [-1]
Output: [-1]

Constraints:

1 <= preorder.length <= 3000
inorder.length == preorder.length
-3000 <= preorder[i], inorder[i] <= 3000
preorder and inorder consist of unique values.
Each value of inorder also appears in preorder.
preorder is guaranteed to be the preorder traversal of the tree.
inorder is guaranteed to be the inorder traversal of the tree.

# The problem statement

here we have the preorder (root - left - right) and the inorder (left - root - right)
from these we need built an binary

from the proeder list the start element is the root element and with this root element from the inorder the left side of the root element is all consider as left side and right as right side

- we will get the root index of the inorder index as the mid and recurse the left till the mid and right from the mid.

- to get the root index, we create an inorder_map with value : index as a key pair.

- to track the preorder index we will use an nonlocal preorder_index variable and increment it.


```python
from typing import Optional, List

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def buildTree(self, preorder: List[int], inorder: List[int]) -> Optional[TreeNode]:
        inorder_map = {value : index for index, value in enumerate(inorder)}
        preorder_index = 0

        def built(left, right):
            nonlocal preorder_index

            if left > right:
                return None
        
            root_val = preorder[preorder_index]  # Fixed: changed () to []
            root = TreeNode(root_val)
            preorder_index += 1

            mid = inorder_map[root_val]

            root.left = built(left, mid-1)
            root.right = built(mid+1,right)

            return root
        
        return built(0, len(inorder)- 1)


# Helper function to print tree in level order for verification
def print_tree(root):
    if not root:
        return "Empty tree"
    
    result = []
    queue = [root]
    
    while queue:
        node = queue.pop(0)
        if node:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            result.append(None)
    
    # Remove trailing None values
    while result and result[-1] is None:
        result.pop()
    
    return result

# Test cases for the buildTree solution
def test_buildTree():
    solution = Solution()
    
    # Test case 1: Basic tree
    # Tree:     3
    #          / \
    #         9   20
    #            /  \
    #           15   7
    preorder1 = [3, 9, 20, 15, 7]
    inorder1 = [9, 3, 15, 20, 7]
    result1 = solution.buildTree(preorder1, inorder1)
    print(f"Test case 1 result: {print_tree(result1)}")
    
    # Test case 2: Single node
    preorder2 = [1]
    inorder2 = [1]
    result2 = solution.buildTree(preorder2, inorder2)
    print(f"Test case 2 result: {print_tree(result2)}")
    
    # Test case 3: Empty tree
    preorder3 = []
    inorder3 = []
    result3 = solution.buildTree(preorder3, inorder3)
    print(f"Test case 3 result: {print_tree(result3)}")
    
    # Test case 4: Left skewed tree
    # Tree: 1
    #        \
    #         2
    #          \
    #           3
    preorder4 = [1, 2, 3]
    inorder4 = [1, 2, 3]
    result4 = solution.buildTree(preorder4, inorder4)
    print(f"Test case 4 result: {print_tree(result4)}")
    
    # Test case 5: Right skewed tree
    # Tree:   3
    #        /
    #       2
    #      /
    #     1
    preorder5 = [3, 2, 1]
    inorder5 = [1, 2, 3]
    result5 = solution.buildTree(preorder5, inorder5)
    print(f"Test case 5 result: {print_tree(result5)}")
        
if __name__ == "__main__":
    test_buildTree()


```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def buildTree(self, preorder: List[int], inorder: List[int]) -> Optional[TreeNode]:
        n = len(preorder)
        indices = list(range(n))
        pDict = dict(zip(preorder, indices))
        iDict = dict(zip(inorder, indices))
        tree = {}
        for val in preorder:
            if not tree:
                tree[val] = TreeNode(val)
            else:
                tVal = preorder[0]
                tree[val] = TreeNode(val)
                while True:
                    if iDict[val] < iDict[tVal]:
                        if not tree[tVal].left:
                            tree[tVal].left = tree[val]
                            break
                        else:
                            tVal = tree[tVal].left.val
                    else:
                        if not tree[tVal].right:
                            tree[tVal].right = tree[val]
                            break
                        else:
                            tVal = tree[tVal].right.val
        return tree[preorder[0]]
        
```
