219. Contains Duplicate II

Given an integer array nums and an integer k, return true if there are two distinct indices i and j in the array such that nums[i] == nums[j] and abs(i - j) <= k.

 

Example 1:

Input: nums = [1,2,3,1], k = 3
Output: true
Example 2:

Input: nums = [1,0,1,1], k = 1
Output: true
Example 3:

Input: nums = [1,2,3,1,2,3], k = 2
Output: false
 

Constraints:

1 <= nums.length <= 105
-109 <= nums[i] <= 109
0 <= k

```python
def containsNearbyDuplicate(nums, k):
    # Using a sliding window approach with a set for O(n) time complexity
    seen = set()
    
    for i in range(len(nums)):
        # If window size exceeds k, remove the leftmost element
        if i > k:
            seen.remove(nums[i - k - 1])
        
        # If current number is already in the set, we found a duplicate within distance k
        if nums[i] in seen:
            return True
        
        # Add current number to the set
        seen.add(nums[i])
    
    return False

print(containsNearbyDuplicate([1,2,3,1], 3))
print(containsNearbyDuplicate([1,2,3,1,2,3], 2))
```

```python
from typing import List

class Solution:
    def containsNearbyDuplicate(self, nums: List[int], k: int) -> bool:
        if len(nums) == len(set(nums)):
            return False
        else:
            for i in range(len(nums)):
                if nums[i] in nums[(i+1):(i+k+1)]:
                    return True
        
            return False

s = Solution()
print(s.containsNearbyDuplicate([1,2,3,2], 3))
```
