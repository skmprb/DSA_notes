Given a positive integer n, write a function that returns the number of set bits in its binary representation (also known as the Hamming weight).

 

Example 1:

Input: n = 11

Output: 3

Explanation:

The input binary string 1011 has a total of three set bits.

Example 2:

Input: n = 128

Output: 1

Explanation:

The input binary string 10000000 has a total of one set bit.

Example 3:

Input: n = 2147483645

Output: 30

Explanation:

The input binary string 1111111111111111111111111111101 has a total of thirty set bits.

 

Constraints:

1 <= n <= 231 - 1
 

```python
class Solution:
    def hammingWeight(self, n: int) -> int:
        count  = 0
        while n:
            n = n & (n-1)
            count += 1
        return count

s = Solution()
print(s.hammingWeight(11))
print(s.hammingWeight(128))
```

```python
class Solution:
    def hammingWeight(self, n: int) -> int:
        # This function counts the number of 1-bits in the binary representation of n
        # It uses Brian Kernighan's algorithm: n & (n-1) removes the rightmost 1-bit
        # What (n-1) does: Subtracting 1 from n flips all the trailing zeros to ones
        # and flips the rightmost 1-bit to 0. For example:
        # n = 12 (1100) -> n-1 = 11 (1011)
        # n = 8 (1000) -> n-1 = 7 (0111)
        # When we do n & (n-1), it removes the rightmost 1-bit from n
        count  = 0
        print(f"Starting with n = {n}, binary: {bin(n)}")
        while n:
            print(f"  Current n: {n} (binary: {bin(n)})")
            n = n & (n-1)
            count += 1
            print(f"  After n & (n-1): {n} (binary: {bin(n)}), count: {count}")
        print(f"Final count: {count}")
        print()
        return count

s = Solution()
print(s.hammingWeight(11))
print(s.hammingWeight(128))
```
