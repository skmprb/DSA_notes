22. Generate Parentheses

Given n pairs of parentheses, write a function to generate all combinations of well-formed parentheses.

 

Example 1:

Input: n = 3
Output: ["((()))","(()())","(())()","()(())","()()()"]
Example 2:

Input: n = 1
Output: ["()"]
 
Constraints:

1 <= n <= 8

# Problem statement

here we can do an recursion/backtracing like,

if the n = 4
the we add the open brackets and check for the close brackets and add till they are equal to the open brackts like we arrange then, till they are equal to the n.


```python
from typing import List
class Solution:
    def generateParenthesis(self, n: int) -> List[str]:

        result = []

        def onePattern(open_count, close_count, current_string):

            if open_count == close_count == n:
                result.append(current_string)
                return
            
            if open_count < n:
                onePattern(open_count + 1, close_count, current_string + "(")

            if close_count < open_count:
                onePattern(open_count, close_count + 1, current_string + ")")
            
        onePattern(0,0,"")
        
        return result

s = Solution()
print(s.generateParenthesis(2))
```

```python
from typing import List
class Solution:
    def generateParenthesis(self, n: int) -> List[str]:

        result = []

        def onePattern(open_count, close_count, current_string):
            # Visualization: Print current state of recursion
            print(f"Current: '{current_string}', Open: {open_count}, Close: {close_count}")

            if open_count == close_count == n:
                # Visualization: Found a complete valid pattern
                print(f"✓ Complete pattern found: '{current_string}'")
                result.append(current_string)
                return
            
            if open_count < n:
                # Visualization: Adding opening parenthesis
                print(f"  → Adding '(' to '{current_string}'")
                onePattern(open_count + 1, close_count, current_string + "(")

            if close_count < open_count:
                # Visualization: Adding closing parenthesis
                print(f"  → Adding ')' to '{current_string}'")
                onePattern(open_count, close_count + 1, current_string + ")")
            
        # Visualization: Starting the pattern generation
        print(f"Starting generation of {n} pairs of parentheses:")
        onePattern(0,0,"")
        
        return result

s = Solution()
print(s.generateParenthesis(2))
```
