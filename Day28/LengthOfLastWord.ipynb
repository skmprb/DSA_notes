## 58. Length of Last Word
### Given a string s consisting of words and spaces, return the length of the last word in the string.

A word is a maximal substring consisting of non-space characters only.

 

Example 1:

Input: s = "Hello World"
Output: 5
Explanation: The last word is "World" with length 5.
Example 2:

Input: s = "   fly me   to   the moon  "
Output: 4
Explanation: The last word is "moon" with length 4.
Example 3:

Input: s = "luffy is still joyboy"
Output: 6
Explanation: The last word is "joyboy" with length 6.
 

Constraints:

1 <= s.length <= 104
s consists of only English letters and spaces ' '.
There will be at least one word in s.

```python
class Solution:
    def lengthOfLastWord(self, s: str) -> int:
        s = s.split()
        return len(s[-1])
    
s = Solution
print(s.lengthOfLastWord(s,"Hello World"))
```

```python
# Time Complexity: O(n) in worst case when the entire string has no spaces or only one word
# Time Complexity: O(k) in best case where k is the length of the last word
# Space Complexity: O(1) - only using a constant amount of extra space for the count variable
def lenLast(s):
    count = 0
    for i in range(len(s)-1,-1,-1):
        if s[i] != " ":
            count+=1
        elif count > 0:
            return count
        

print(lenLast("   fly me   to   the moon  "))
print(lenLast("Hello World"))
print(lenLast("luffy is still joyboy"))
```

```python
# Optimized Solution 1: Using strip() + split()
# Time Complexity: O(n) - strip() and split() both traverse the string
# Space Complexity: O(1) - only stores the last word after split
def lengthOfLastWordOptimized(s):
    return len(s.strip().split()[-1])

print(lengthOfLastWordOptimized("   fly me   to   the moon  "))
print(lengthOfLastWordOptimized("Hello World"))
print(lengthOfLastWordOptimized("luffy is still joyboy"))
print(lengthOfLastWordOptimized("a "))
```

```python
# Optimized Solution 2: Using rsplit() with maxsplit
# Time Complexity: O(k) - only processes from right until first word found
# Space Complexity: O(1) - only stores the last word
def lengthOfLastWordRsplit(s):
    return len(s.rsplit(maxsplit=1)[-1])

print(lengthOfLastWordRsplit("   fly me   to   the moon  "))
print(lengthOfLastWordRsplit("Hello World"))
print(lengthOfLastWordRsplit("luffy is still joyboy"))
```

```python
# Optimized Solution 3: Reverse iteration with early termination
# Time Complexity: O(k) - k is length of last word + trailing spaces
# Space Complexity: O(1) - no extra space used
def lengthOfLastWordReverse(s):
    length = 0
    i = len(s) - 1
    
    while i >= 0 and s[i] == ' ':
        i -= 1
    
    while i >= 0 and s[i] != ' ':
        length += 1
        i -= 1
    
    return length

print(lengthOfLastWordReverse("   fly me   to   the moon  "))
print(lengthOfLastWordReverse("Hello World"))
print(lengthOfLastWordReverse("luffy is still joyboy"))
```

### Comparison:
- **Solution 1 (strip + split)**: Clean and Pythonic, good for most cases
- **Solution 2 (rsplit)**: Most efficient - only splits once from right, best performance
- **Solution 3 (reverse iteration)**: True O(k) with no string splitting overhead, best for very long strings
