## Part 1: Understanding Recursion Basics

### What is Recursion?

**Recursion** is when a function calls itself to solve a problem by breaking it down into smaller, similar subproblems.

**Key Components:**
1. **Base Case**: The stopping condition (prevents infinite recursion)
2. **Recursive Case**: The function calling itself with a smaller/simpler input
3. **Progress**: Each recursive call must move toward the base case

### Example 1: Simple Countdown (Understanding the Flow)

````python
def countdown(n):
    # Base Case: When to STOP
    if n == 0:
        print("Done!")
        return
    
    # Recursive Case: Do something, then call yourself with smaller input
    print(f"Counting: {n}")
    countdown(n - 1)  # Call itself with n-1

# Test it
countdown(5)

````

**What's happening step by step:**

```
countdown(5) → prints 5, calls countdown(4)
  countdown(4) → prints 4, calls countdown(3)
    countdown(3) → prints 3, calls countdown(2)
      countdown(2) → prints 2, calls countdown(1)
        countdown(1) → prints 1, calls countdown(0)
          countdown(0) → prints "Done!", RETURNS (base case)
        countdown(1) → finishes, RETURNS
      countdown(2) → finishes, RETURNS
    countdown(3) → finishes, RETURNS
  countdown(4) → finishes, RETURNS
countdown(5) → finishes, RETURNS
```

### Example 2: Factorial (Returning Values)

````python
def factorial(n):
    # Base Case: factorial of 0 or 1 is 1
    if n <= 1:
        print(f"Base case reached: factorial({n}) = 1")
        return 1
    
    # Recursive Case: n! = n * (n-1)!
    print(f"Calculating factorial({n}) = {n} * factorial({n-1})")
    result = n * factorial(n - 1)
    print(f"Returning factorial({n}) = {result}")
    return result

# Test it
print("\nFinal result:", factorial(5))
````

**Detailed Execution Flow:**

```
factorial(5)
  → needs 5 * factorial(4)
  → waits for factorial(4) to return
  
    factorial(4)
      → needs 4 * factorial(3)
      → waits for factorial(3) to return
      
        factorial(3)
          → needs 3 * factorial(2)
          → waits for factorial(2) to return
          
            factorial(2)
              → needs 2 * factorial(1)
              → waits for factorial(1) to return
              
                factorial(1)
                  → BASE CASE! Returns 1
                  
              factorial(2) gets 1, calculates 2 * 1 = 2, returns 2
              
          factorial(3) gets 2, calculates 3 * 2 = 6, returns 6
          
      factorial(4) gets 6, calculates 4 * 6 = 24, returns 24
      
  factorial(5) gets 24, calculates 5 * 24 = 120, returns 120
```

**Key Insight:** Each function call WAITS for its child call to return before it can complete!

### Example 3: Visualizing the Call Stack

````python
# Let's track the call stack manually
call_stack = []

def factorial_with_stack(n):
    # Add to stack when entering function
    call_stack.append(f"factorial({n})")
    print(f"PUSH: {call_stack}")
    
    # Base Case
    if n <= 1:
        call_stack.pop()
        print(f"POP (base case): {call_stack}")
        return 1
    
    # Recursive Case
    result = n * factorial_with_stack(n - 1)
    
    # Remove from stack when returning
    call_stack.pop()
    print(f"POP (returning {result}): {call_stack}")
    return result

# Test it
print("\nFinal result:", factorial_with_stack(5))
````

### Example 4: Sum of Numbers (Simple Practice)

````python
def sum_numbers(n):
    """
    Calculate sum of numbers from 1 to n
    Example: sum_numbers(5) = 1 + 2 + 3 + 4 + 5 = 15
    """
    # Base Case: sum of 0 is 0
    if n == 0:
        return 0
    
    # Recursive Case: n + sum of (n-1)
    return n + sum_numbers(n - 1)

# Test it
print("Sum of 1 to 5:", sum_numbers(5))
print("Sum of 1 to 10:", sum_numbers(10))
````

**Trace for sum_numbers(5):**

```
sum_numbers(5) = 5 + sum_numbers(4)
                   = 5 + (4 + sum_numbers(3))
                   = 5 + (4 + (3 + sum_numbers(2)))
                   = 5 + (4 + (3 + (2 + sum_numbers(1))))
                   = 5 + (4 + (3 + (2 + (1 + sum_numbers(0)))))
                   = 5 + (4 + (3 + (2 + (1 + 0))))
                   = 5 + (4 + (3 + (2 + 1)))
                   = 5 + (4 + (3 + 3))
                   = 5 + (4 + 6)
                   = 5 + 10
                   = 15
```

### Key Takeaways from Part 1:

1. **Every recursive function needs a base case** - otherwise it runs forever!
2. **Each call waits for its child calls to complete** - this creates the call stack
3. **Functions return in reverse order** - last called, first returned (LIFO - Last In First Out)
4. **Each function call is independent** - has its own variables and parameters
5. **Progress toward base case is essential** - each call should get closer to stopping

````python
# Problem 1: Write a recursive function to calculate power (x^n)
def power(x, n):
    # Your code here
    if n < 1:
        return 1
    
    result = x * power(x,n-1)
    return result

# Test: power(2, 5) should return 32
print(power(2, 5))
````

````python
# Problem 2: Write a recursive function to count digits in a number
def count_digits(n):
    # Your code here
    if n <= 0:
        return 0
    
    result = 1+count_digits(n//10)
    return result

# Test: count_digits(12345) should return 5
print(count_digits(1234567))
````

````python
# Problem 3: Write a recursive function to print numbers from n to 1
def print_reverse(n):
    # Your code here
    # Base case: when n reaches 1, print it and stop recursion
    if n <= 1:
        print(1)
        return
    
    print(n)
    print_reverse(n-1)

# Alternative approach 2: Recursive with different base case structure
def print_reverse_alt(n):
    if n < 1:
        return
    print(n)
    print_reverse_alt(n-1)

# Test: print_reverse(5) should print 5 4 3 2 1
print_reverse(5)
print_reverse_alt(5)
````
