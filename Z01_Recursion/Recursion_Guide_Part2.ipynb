### 1. Multiple Recursive Calls

Some problems require calling the function multiple times within one call.

#### Example: Fibonacci Numbers

````python
def fibonacci(n):
    """
    Fibonacci sequence: 0, 1, 1, 2, 3, 5, 8, 13, 21...
    fib(n) = fib(n-1) + fib(n-2)
    """
    # Base Cases
    if n == 0:
        return 0
    if n == 1:
        return 1
    
    # Recursive Case: TWO recursive calls!
    return fibonacci(n - 1) + fibonacci(n - 2)

# Test it
print("Fibonacci numbers:")
for i in range(8):
    print(f"fib({i}) = {fibonacci(i)}")
````

**Detailed Execution Tree for fibonacci(5):**

```
                        fib(5)
                       /      \
                   fib(4)      fib(3)
                   /    \      /    \
               fib(3)  fib(2) fib(2) fib(1)
               /   \    /  \   /  \     |
           fib(2) fib(1) fib(1) fib(0) fib(1) fib(0)  1
           /   \    |      |      |      |      |
       fib(1) fib(0) 1     1      0      1      0
         |      |
         1      0
```

**Key Points:**
- Each call spawns TWO more calls (except base cases)
- Creates a tree structure, not a linear chain
- Notice: fib(3) is calculated TWICE! (inefficient - we'll fix this later)

#### Step-by-Step Execution Order:

````python
# Let's trace the execution order
call_count = 0

def fibonacci_traced(n, indent=0):
    global call_count
    call_count += 1
    call_num = call_count
    
    print(" " * indent + f"[{call_num}] Calling fib({n})")
    
    # Base Cases
    if n == 0:
        print(" " * indent + f"[{call_num}] Base case: fib(0) = 0")
        return 0
    if n == 1:
        print(" " * indent + f"[{call_num}] Base case: fib(1) = 1")
        return 1
    
    # Recursive Case
    print(" " * indent + f"[{call_num}] Need fib({n-1}) + fib({n-2})")
    
    left = fibonacci_traced(n - 1, indent + 2)
    print(" " * indent + f"[{call_num}] Got fib({n-1}) = {left}")
    
    right = fibonacci_traced(n - 2, indent + 2)
    print(" " * indent + f"[{call_num}] Got fib({n-2}) = {right}")
    
    result = left + right
    print(" " * indent + f"[{call_num}] Returning fib({n}) = {result}")
    return result

# Test with small number
call_count = 0
print("\nFinal result:", fibonacci_traced(4))
print(f"\nTotal function calls: {call_count}")
````

**Important Observation:**

Notice how the function:
1. Calls fib(n-1) FIRST and WAITS for it to complete
2. THEN calls fib(n-2) and waits for it
3. THEN adds the results and returns

This is **left-to-right evaluation** - the left recursive call completes entirely before the right one starts!

### 2. Helper Functions with Extra Parameters

Sometimes we need extra information that wasn't in the original problem. We use helper functions for this.

#### Example: Print Array Elements Recursively

````python
def print_array(arr):
    """
    Main function - user calls this
    """
    def helper(arr, index):
        """
        Helper function - has extra parameter 'index'
        """
        # Base Case: reached end of array
        if index >= len(arr):
            return
        
        # Print current element
        print(f"arr[{index}] = {arr[index]}")
        
        # Recursive Case: move to next index
        helper(arr, index + 1)
    
    # Start from index 0
    helper(arr, 0)

# Test it
print_array([10, 20, 30, 40, 50])
````

**Why use a helper function?**

- User only provides the array: `print_array([10, 20, 30])`
- Helper needs index to track position: `helper(arr, index)`
- Keeps the main function interface clean

**Execution Flow:**
```
print_array([10, 20, 30])
  → helper([10, 20, 30], 0)
      → prints arr[0] = 10
      → helper([10, 20, 30], 1)
          → prints arr[1] = 20
          → helper([10, 20, 30], 2)
              → prints arr[2] = 30
              → helper([10, 20, 30], 3)
                  → index >= len(arr), BASE CASE, returns
              → returns
          → returns
      → returns
```

#### Example: Sum of Array with Helper

````python
def sum_array(arr):
    def helper(arr, index):
        # Base Case: reached end
        if index >= len(arr):
            return 0
        
        # Recursive Case: current element + sum of rest
        return arr[index] + helper(arr, index + 1)
    
    return helper(arr, 0)

# Test it
print("Sum:", sum_array([1, 2, 3, 4, 5]))  # Should be 15
````

### 3. Accumulator Pattern

Instead of building the result on the way back, we can build it on the way down using an accumulator.

#### Example: Factorial with Accumulator

````python
# Traditional recursion (builds result on the way BACK)
def factorial_traditional(n):
    if n <= 1:
        return 1
    return n * factorial_traditional(n - 1)

# Accumulator pattern (builds result on the way DOWN)
def factorial_accumulator(n, acc=1):
    """
    acc = accumulator, stores the result so far
    """
    print(f"factorial_accumulator({n}, acc={acc})")
    
    # Base Case
    if n <= 1:
        print(f"Base case reached, returning {acc}")
        return acc
    
    # Recursive Case: pass the accumulated result down
    return factorial_accumulator(n - 1, acc * n)

# Test both
print("Traditional:")
print(factorial_traditional(5))

print("\nAccumulator:")
print(factorial_accumulator(5))
````

**Comparison:**

**Traditional (builds on return):**
```
factorial(5)
  → 5 * factorial(4)
      → 5 * (4 * factorial(3))
          → 5 * (4 * (3 * factorial(2)))
              → 5 * (4 * (3 * (2 * factorial(1))))
                  → 5 * (4 * (3 * (2 * 1)))
                  → 5 * (4 * (3 * 2))
                  → 5 * (4 * 6)
                  → 5 * 24
                  → 120
```

**Accumulator (builds going down):**
```
factorial_accumulator(5, 1)
  → factorial_accumulator(4, 5)    # acc = 1 * 5 = 5
      → factorial_accumulator(3, 20)   # acc = 5 * 4 = 20
          → factorial_accumulator(2, 60)   # acc = 20 * 3 = 60
              → factorial_accumulator(1, 120)  # acc = 60 * 2 = 120
                  → returns 120 (base case)
```

**Advantage:** Accumulator pattern is more memory efficient (tail recursion optimization possible)

### 4. Returning Multiple Values (Tuples)

Sometimes we need to return more than one piece of information from a recursive call.

#### Example: Find Min and Max in Array

````python
def find_min_max(arr):
    def helper(arr, index):
        # Base Case: last element
        if index == len(arr) - 1:
            return (arr[index], arr[index])  # (min, max)
        
        # Recursive Case: get min/max from rest of array
        rest_min, rest_max = helper(arr, index + 1)
        
        # Compare current element with rest
        current_min = min(arr[index], rest_min)
        current_max = max(arr[index], rest_max)
        
        return (current_min, current_max)
    
    return helper(arr, 0)

# Test it
arr = [5, 2, 9, 1, 7, 3]
min_val, max_val = find_min_max(arr)
print(f"Array: {arr}")
print(f"Min: {min_val}, Max: {max_val}")
````

**Execution Trace for [5, 2, 9]:**

```
helper([5, 2, 9], 0)
  → needs helper([5, 2, 9], 1)
  
      helper([5, 2, 9], 1)
        → needs helper([5, 2, 9], 2)
        
            helper([5, 2, 9], 2)
              → BASE CASE: returns (9, 9)
              
        → gets (9, 9)
        → compares arr[1]=2 with (9, 9)
        → returns (min(2,9), max(2,9)) = (2, 9)
        
  → gets (2, 9)
  → compares arr[0]=5 with (2, 9)
  → returns (min(5,2), max(5,9)) = (2, 9)
```

**Key Point:** We use tuple unpacking to get both values:
```python
rest_min, rest_max = helper(arr, index + 1)
```

#### Example: Count Positive and Negative Numbers

````python
def count_pos_neg(arr):
    def helper(arr, index):
        # Base Case: empty array
        if index >= len(arr):
            return (0, 0)  # (positive_count, negative_count)
        
        # Get counts from rest of array
        pos_count, neg_count = helper(arr, index + 1)
        
        # Update counts based on current element
        if arr[index] > 0:
            pos_count += 1
        elif arr[index] < 0:
            neg_count += 1
        
        return (pos_count, neg_count)
    
    return helper(arr, 0)

# Test it
arr = [5, -2, 3, -7, 0, 8, -1]
positive, negative = count_pos_neg(arr)
print(f"Array: {arr}")
print(f"Positive: {positive}, Negative: {negative}")
````

### 5. Real Example: This is How Your Tree Code Works!

Now let's see how this applies to your tree height and balance problem:

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def is_balanced(root):
    def helper(node):
        print(f"\nCalling helper on node: {node.val if node else 'None'}")
        
        # Base Case: empty node
        if not node:
            print("  → Base case: returning (height=0, balanced=True)")
            return (0, True)
        
        # Recursive Case: get info from BOTH children
        print(f"  → Getting left subtree info...")
        left_height, left_balanced = helper(node.left)
        print(f"  → Left subtree: height={left_height}, balanced={left_balanced}")
        
        print(f"  → Getting right subtree info...")
        right_height, right_balanced = helper(node.right)
        print(f"  → Right subtree: height={right_height}, balanced={right_balanced}")
        
        # Calculate current node's height
        height = max(left_height, right_height) + 1
        
        # Check if current node is balanced
        is_balanced = (left_balanced and 
                      right_balanced and 
                      abs(left_height - right_height) <= 1)
        
        print(f"  → Node {node.val}: height={height}, balanced={is_balanced}")
        return (height, is_balanced)
    
    return helper(root)

# Create a simple tree
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)

height, balanced = is_balanced(root)
print(f"\n{'='*50}")
print(f"Final Result: Height={height}, Balanced={balanced}")
````

**What's happening:**

1. **Multiple recursive calls**: `helper(node.left)` and `helper(node.right)`
2. **Returning multiple values**: `(height, is_balanced)` tuple
3. **Tuple unpacking**: `left_height, left_balanced = helper(node.left)`
4. **Building result from children**: Uses left and right info to calculate current node's info

This combines ALL the concepts from Part 2!

### Key Takeaways from Part 2:

1. **Multiple recursive calls** create tree-like execution patterns (not linear)
2. **Helper functions** let you add extra parameters without changing the main interface
3. **Accumulator pattern** builds results going down instead of coming back up
4. **Returning tuples** lets you return multiple pieces of information
5. **Tuple unpacking** extracts multiple return values: `a, b = function()`

````python
# Problem 1: Write a recursive function to find if a number exists in an array
def find_number(arr, target):
    # Use a helper function with index parameter
    # Your code here

    def helper(arr, idx):
        # Base case: if we've reached the end of the array, target not found
        if idx == len(arr):
            return False
        
        # If current element matches target, return True
        if arr[idx] == target:
            return True
        
        # Otherwise, continue searching in the rest of the array
        return helper(arr, idx + 1)
        
    return helper(arr, 0)

print(find_number([1, 2, 3, 4, 5], 3))

# Test: find_number([1, 2, 3, 4, 5], 3) should return True
````

````python
# Problem 2: Write a recursive function to reverse a string
def reverse_string(s):
    # Hint: reverse("hello") = reverse("ello") + "h"
    # Your code here
    
    # This code defines a nested helper function called 'reverse'
    # The helper function implements the core recursive logic for string reversal
    def reverse(st):
        # Base case: if the string is empty, return an empty string
        # This stops the recursion when we've processed all characters
        if not st:
            return ""
        # Recursive case: take the first character and append it to the end
        # of the reversed substring (everything except the first character)
        # This effectively moves each character from front to back
        return reverse(st[1:]) + st[0]

    # The main function uses the same logic as the helper function
    # It calls reverse on everything except the first character (s[1:])
    # Then appends the first character (s[0]) to the end
    # The conditional handles the edge case of an empty string
    return reverse(s[1:]) + s[0] if s else ""

print(reverse_string("why this recurssion is this mauch tricky"))
# Test: reverse_string("hello") should return "olleh"
````

````python
# Problem 3: Write a recursive function that returns both sum and product of array
def sum_and_product(arr):
    # Return a tuple (sum, product)
    # Your code here
    def helper(arr, idx, sum, prod):

        # Base case: when we've processed all elements, return the accumulated sum and product
        if idx == len(arr):
            return (sum, prod)
        
        # Process current element before recursion (add to sum, multiply to product)
        sum += arr[idx]
        prod *= arr[idx]
        
        # Recursive call with updated sum and product, and next index
        return helper(arr, idx + 1, sum, prod)

    # Start recursion with index 0, sum 0, and product 1
    return helper(arr, 0, 0, 1)
            
print(sum_and_product([1, 2, 3, 4]))
# Test: sum_and_product([1, 2, 3, 4]) should return (10, 24)
````

````python
# Problem 4: Write a recursive function to calculate sum using accumulator pattern
def sum_accumulator(n, acc=0):
    # Calculate sum of 1 to n using accumulator
    # Your code here
    if n <= 0:
        return acc

    return sum_accumulator(n-1,n+acc)

print(sum_accumulator(5))
# Test: sum_accumulator(5) should return 15
````
