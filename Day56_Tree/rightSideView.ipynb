199. Binary Tree Right Side View

Given the root of a binary tree, imagine yourself standing on the right side of it, return the values of the nodes you can see ordered from top to bottom.

 

Example 1:

- Input: root = [1,2,3,null,5,null,4]
- Output: [1,3,4]

Example 2:

- Input: root = [1,2,3,4,null,null,null,5]
- Output: [1,3,4,5]

Example 3:

- Input: root = [1,null,3]
- Output: [1,3]

Example 4:

- Input: root = []
- Output: []

Constraints:

The number of nodes in the tree is in the range [0, 100].
-100 <= Node.val <= 100

# Problem Satatement

here there is a tree and i have to return the elements that are last elements in a level.

for example there are levels: 1, 2, 3 top to bottom
- i have to return the last element from level 1 list of elements.
- here the top will always be root, since it only one elemet
- even in a level the element is start of the list, if it has not other elements exist we have to return that.
- if no root then we return empty list.

here we will be using the DFS first,

- we will check the right-most element first in a level, if not root.right and we go to the root.left and its  same for root.right.right, root.left.left.
- we can track the depth/levels by using the helper function with root and depth.
- here we will be doing recursion with right most element first and add the val to result, so here if the right most is not then it will go for the left.
- and also we will check the length og the result with depth once the lement from the one depth/level added it will wait till next level / depth







```python
# Depth first Approach

from typing import Optional, List
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def rightSideView(self, root: Optional[TreeNode]) -> List[int]:
        res = []
        def dfs(root, depth):
            if not root:
                return None
            if len(res) == depth:
                res.append(root.val)
            
            # important to call the right element first.
            dfs(root.right, depth+1)
            dfs(root.left, depth+1)
        dfs(root, 0)
        return res


# Test case 1: Original test case
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.left.left = TreeNode(5)

s = Solution()
print(s.rightSideView(root))

# Test case 2: Empty tree
root2 = None
print(s.rightSideView(root2))

# Test case 3: Single node
root3 = TreeNode(1)
print(s.rightSideView(root3))

# Test case 4: Only left children
root4 = TreeNode(1)
root4.left = TreeNode(2)
root4.left.left = TreeNode(3)
print(s.rightSideView(root4))

# Test case 5: Only right children
root5 = TreeNode(1)
root5.right = TreeNode(2)
root5.right.right = TreeNode(3)
print(s.rightSideView(root5))

# Test case 6: Complete binary tree
root6 = TreeNode(1)
root6.left = TreeNode(2)
root6.right = TreeNode(3)
root6.left.left = TreeNode(4)
root6.left.right = TreeNode(5)
root6.right.left = TreeNode(6)
root6.right.right = TreeNode(7)
print(s.rightSideView(root6))

# Test case 7: Unbalanced tree with mixed structure
root7 = TreeNode(1)
root7.left = TreeNode(2)
root7.right = TreeNode(3)
root7.left.right = TreeNode(5)
root7.right.right = TreeNode(4)
print(s.rightSideView(root7))
```

```python
# Bredth first approach
from typing import Optional, List
from collections import deque
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def rightSideView(self, root: Optional[TreeNode]) -> List[int]:
        if not root:
            return []
        
        res = []
        queue = deque([root])

        while queue:
            level_size = len(queue)

            for i in range(level_size):
                node = queue.popleft()

                if i == level_size - 1:
                    res.append(node.val)
                
                if node.left:
                    queue.append(node.left)
                if node.right:
                    queue.append(node.right)
        
        return res

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.left.left = TreeNode(5)

s = Solution()
print(s.rightSideView(root))
```
