# 🧮 Expression Evaluator — Binary Tree

## What You're Building
A math expression parser and evaluator (like compilers do):
- Parse expression string into a tree
- Evaluate the tree to get result
- Support: +, -, *, /, parentheses

## Why Binary Tree?
- Each operator has exactly 2 operands (left and right)
- Tree structure naturally handles operator precedence
- Evaluation = post-order traversal (evaluate children first, then apply operator)

```
Expression: (2 + 3) * 4

Tree:
        *
       / \
      +   4
     / \
    2   3

Post-order: 2, 3, +, 4, * → evaluate bottom-up → 5 * 4 = 20
```
