# 📁 File System — N-ary Tree

## What You're Building
A file system like Linux/Windows where:
- Folders can contain folders and files (N children)
- Navigate with `cd`, list with `ls`, show path with `pwd`
- Create/delete files and folders

## Why N-ary Tree?
- Each folder can have **any number** of children (not just 2)
- Natural parent-child hierarchy
- Path = root to current node traversal

```
         / (root)
       / |  \
    home  etc  var
    /  \       |
 user1 user2  log
   |          |
 docs      app.log
```
