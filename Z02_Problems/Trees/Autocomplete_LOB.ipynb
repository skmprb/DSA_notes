# 🔍 Autocomplete Engine — Trie (Prefix Tree)

## What You're Building
A search autocomplete system (like Google search bar):
- Insert words into dictionary
- Type a prefix → get all matching suggestions
- Track frequency → suggest most popular first

## Why Trie?
- Prefix-based lookup in O(prefix_length) — not O(n) scanning all words
- Shared prefixes save memory ("cat", "car", "card" share "ca")
- Natural for character-by-character typing

```
         (root)
        /     \
       c       d
      / \       \
     a   o      o
    / \   \      \
   t   r   d     g
       |   e
       d

Words: cat, car, card, code, dog
```
