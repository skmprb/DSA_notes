# 📦 Huffman Compression — Binary Tree

## What You're Building
A file compression algorithm (used in ZIP, GZIP, JPEG):
- Count character frequencies
- Build Huffman tree (frequent chars get shorter codes)
- Encode text → binary string
- Decode binary string → original text

## Why Binary Tree?
- Left branch = 0, Right branch = 1
- Path from root to leaf = binary code for that character
- Frequent chars are near root (short path = short code)
- Rare chars are deep (long path = long code)

```
Text: "aaabbc"
Frequencies: a=3, b=2, c=1

Huffman Tree:
        (6)
       /   \
     a(3)  (3)
           / \
         b(2) c(1)

Codes: a=0, b=10, c=11
Encoded: 000101011 (9 bits vs 48 bits uncompressed!)
```

## Phase 3: Build Huffman Tree

**Algorithm (uses a min-heap):**
1. Create a leaf node for each character, push to min-heap
2. While heap has more than 1 node:
   - Pop two smallest frequency nodes
   - Create parent with freq = sum of both
   - Left child = smaller, Right child = larger
   - Push parent back to heap
3. Last remaining node = root of Huffman tree
