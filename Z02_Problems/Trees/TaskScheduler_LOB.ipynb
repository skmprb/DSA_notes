# ⚡ Task Scheduler — Heap (Priority Queue)

## What You're Building
A priority-based task scheduler (like OS task manager):
- Add tasks with priority levels
- Always execute highest priority task first
- Support: add task, execute next, peek, change priority

## Why Heap (Binary Tree)?
- Get max/min priority in O(1)
- Insert new task in O(log n)
- Remove highest priority in O(log n)
- A heap IS a binary tree stored in an array!

```
Max-Heap (highest priority at top):

        [Priority 10: "Critical Bug"]
       /                            \
[P7: "Feature"]              [P8: "Security Patch"]
    /       \                      /
[P3: "Docs"] [P5: "Refactor"] [P2: "Typo"]

Array: [(10,Bug), (7,Feature), (8,Security), (3,Docs), (5,Refactor), (2,Typo)]
```
