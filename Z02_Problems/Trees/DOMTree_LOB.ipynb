# 🌐 DOM Tree — HTML Rendering Engine

## What You're Building
A simplified DOM (Document Object Model) like browsers use to render HTML:
- Parse HTML tags into a tree
- Query elements (getElementById, getElementsByTag)
- Render tree as indented HTML
- Traverse: BFS (level-order) and DFS (depth-first)

## Why Tree?
- HTML is inherently nested: `<div>` contains `<p>` contains `<span>`
- Parent-child relationships map directly to tree nodes
- Rendering = tree traversal with indentation

```html
<html>
  <head>
    <title>My Page</title>
  </head>
  <body>
    <div id="main">
      <h1>Hello</h1>
      <p class="intro">Welcome</p>
    </div>
  </body>
</html>
```

```
Tree:
        html
       /    \
    head    body
     |       |
   title    div#main
            /    \
          h1    p.intro
```
