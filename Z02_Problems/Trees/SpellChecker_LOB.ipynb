## What You're Building
A spell checker that:
- Checks if a word exists in dictionary
- Suggests corrections for misspelled words
- Handles: missing letter, extra letter, wrong letter, swapped letters

## Why Trie?
- O(word_length) lookup — faster than scanning entire dictionary
- Can explore "nearby" words by branching at each level
- Prefix sharing = memory efficient for large dictionaries

```
User types: "helo"
Suggestions: "hello", "help", "hero", "halo"

How? Trie lets us explore words within 1-2 edits efficiently.
```
