# ⚙️ OS Process Scheduling — Round Robin with Circular Linked List

## What You're Building
A CPU scheduler that gives each process a fixed time slice (quantum), then moves to the next.
- Add process to scheduler
- Execute current process for 1 quantum
- If process finishes, remove it
- If not, move to next process
- Keep cycling until all processes are done

## Why Circular Linked List?
- After the last process, CPU goes back to the first — endless cycle
- Processes can be added/removed dynamically without shifting
- Fair scheduling: every process gets equal CPU time
