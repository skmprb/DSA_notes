# 💾 Memory Allocation — Free Block Linked List

## What You're Building
A simplified OS memory allocator that manages free memory blocks.
- Allocate memory (find a free block big enough)
- Free memory (return block to free list)
- Merge adjacent free blocks (defragmentation)

## Why Linked List?
- Free blocks are scattered in memory — not contiguous
- Blocks are created/destroyed dynamically
- Need to traverse and find "best fit" or "first fit" block
- Merging adjacent blocks = merging adjacent nodes

```
Memory: [USED 0-99] [FREE 100-299] [USED 300-499] [FREE 500-799] [USED 800-1023]

Free List: [100-299, size=200] → [500-799, size=300] → None
```
