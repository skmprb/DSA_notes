# 🗄️ LRU Cache — Doubly Linked List + HashMap

## What You're Building
A cache with fixed capacity that evicts the **Least Recently Used** item when full.
Used in: databases, CDNs, browser cache, DNS cache.

- `get(key)` — return value if exists, mark as recently used
- `put(key, value)` — insert/update, evict LRU if full

## Why DLL + HashMap?
- **HashMap** → O(1) lookup by key
- **DLL** → O(1) move-to-front (most recent) and remove-from-tail (least recent)
- Together → O(1) for both get and put!

```
Most Recent                          Least Recent
  HEAD <-> [D] <-> [B] <-> [A] <-> [C] <-> TAIL
                                             ^
                                     evict this first
```
