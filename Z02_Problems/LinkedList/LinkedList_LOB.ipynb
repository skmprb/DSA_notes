# 🔗 Linked List — Learn On Build (LOB)

## Real-World App: Browser History Navigation

We'll build a **Browser History** system step by step, learning linked list concepts as we go.

**Why Linked List here?**
- Dynamic size (unknown number of pages)
- O(1) insert/delete (vs O(n) array shifting)
- Natural prev/next navigation

---

## Phase 1: The Node — Building Block

Every linked list starts with a **Node**. Think of it as a single browser tab storing:
- The URL (data)
- A pointer to the next page
- A pointer to the previous page

````python
class Page:
    def __init__(self, url):
        self.url = url
        self.prev = None
        self.next = None

# Test: create a single page
p = Page("google.com")
print(f"URL: {p.url}, Prev: {p.prev}, Next: {p.next}")
````

## Phase 2: Visiting Pages — Insertion at Tail

When you visit a new page, it gets **appended after the current page**.

```
Before: [Google] <-> [Amazon]   (current = Amazon)
After:  [Google] <-> [Amazon] <-> [YouTube]   (current = YouTube)
```

**Linked List Concept:** Insert after a given node — O(1)

````python
class BrowserHistory:
    def __init__(self, homepage):
        self.current = Page(homepage)
    
    def visit(self, url):
        """Visit a new page — discard forward history"""
        new_page = Page(url)
        self.current.next = new_page
        new_page.prev = self.current
        self.current = new_page
    
    def current_page(self):
        return self.current.url

# Test
browser = BrowserHistory("google.com")
browser.visit("amazon.com")
browser.visit("youtube.com")
print(f"Current: {browser.current_page()}")  # youtube.com
````

## Phase 3: Back Button — Traversal Backward

Going back = move `current` pointer to `current.prev`

**Linked List Concept:** Traversal using pointers — O(1) per step

````python
class BrowserHistory:
    def __init__(self, homepage):
        self.current = Page(homepage)
    
    def visit(self, url):
        new_page = Page(url)
        self.current.next = new_page
        new_page.prev = self.current
        self.current = new_page
    
    def back(self, steps):
        """Go back 'steps' pages"""
        while steps > 0 and self.current.prev:
            self.current = self.current.prev
            steps -= 1
        return self.current.url
    
    def forward(self, steps):
        """Go forward 'steps' pages"""
        while steps > 0 and self.current.next:
            self.current = self.current.next
            steps -= 1
        return self.current.url

# Test
browser = BrowserHistory("google.com")
browser.visit("amazon.com")
browser.visit("youtube.com")
browser.visit("github.com")

print(browser.back(2))     # amazon.com
print(browser.forward(1))  # youtube.com
````

## Phase 4: Visit After Going Back — Deletion of Forward History

When you go back and visit a new page, all forward history is **discarded**.

```
Before: [Google] <-> [Amazon] <-> [YouTube] <-> [GitHub]
                       ^ current
Visit "netflix.com":
After:  [Google] <-> [Amazon] <-> [Netflix]
                                    ^ current
```

**Linked List Concept:** Deleting all nodes after a given node — O(1) (just cut the link, garbage collector handles the rest)

````python
# Already handled! When we do:
#   self.current.next = new_page
# The old forward chain loses its reference and gets garbage collected.

# Full test:
browser = BrowserHistory("google.com")
browser.visit("amazon.com")
browser.visit("youtube.com")
browser.visit("github.com")

print(f"At: {browser.back(2)}")  # amazon.com

# Now visit netflix — youtube & github are gone
browser.visit("netflix.com")
print(f"At: {browser.current.url}")  # netflix.com
print(f"Forward exists? {browser.current.next}")  # None — forward history wiped
````

## Phase 5: Display Full History — Full Traversal

**Linked List Concept:** Traverse from head to tail, printing each node.

````python
class BrowserHistory:
    def __init__(self, homepage):
        self.current = Page(homepage)
    
    def visit(self, url):
        new_page = Page(url)
        self.current.next = new_page
        new_page.prev = self.current
        self.current = new_page
    
    def back(self, steps):
        while steps > 0 and self.current.prev:
            self.current = self.current.prev
            steps -= 1
        return self.current.url
    
    def forward(self, steps):
        while steps > 0 and self.current.next:
            self.current = self.current.next
            steps -= 1
        return self.current.url
    
    def display_history(self):
        """Traverse to head, then print all pages"""
        node = self.current
        while node.prev:
            node = node.prev
        
        pages = []
        while node:
            marker = " 👈" if node == self.current else ""
            pages.append(f"[{node.url}{marker}]")
            node = node.next
        print(" <-> ".join(pages))

# Full demo
browser = BrowserHistory("google.com")
browser.visit("amazon.com")
browser.visit("youtube.com")
browser.visit("github.com")
browser.display_history()

browser.back(2)
browser.display_history()

browser.visit("netflix.com")
browser.display_history()
````
