# ↩️ Undo/Redo System — Doubly Linked List

## What You're Building
A text editor's undo/redo system (like VS Code, Photoshop).
- Perform actions (type, delete, format)
- Undo — go back to previous state
- Redo — go forward to undone state
- New action after undo — discard redo history

## Why Doubly Linked List?
- Each state points to previous AND next state
- Undo = move backward, Redo = move forward
- New action after undo = cut the forward chain (O(1))
