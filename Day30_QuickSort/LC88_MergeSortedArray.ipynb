## 📋 Problem Statement

You are given two integer arrays `nums1` and `nums2`, sorted in **non-decreasing order**, and two integers `m` and `n`, representing the number of elements in `nums1` and `nums2` respectively.

**Merge** `nums2` into `nums1` as one sorted array.

The final sorted array should not be returned by the function, but instead be **stored inside the array nums1**. 

To accommodate this, `nums1` has a length of `m + n`, where:
- The first `m` elements denote the elements that should be merged
- The last `n` elements are set to `0` and should be ignored
- `nums2` has a length of `n`

### Examples:

**Example 1:**
```
Input: nums1 = [1,2,3,0,0,0], m = 3, nums2 = [2,5,6], n = 3
Output: [1,2,2,3,5,6]
Explanation: Merge [1,2,3] and [2,5,6] → [1,2,2,3,5,6]
```

**Example 2:**
```
Input: nums1 = [1], m = 1, nums2 = [], n = 0
Output: [1]
```

**Example 3:**
```
Input: nums1 = [0], m = 0, nums2 = [1], n = 1
Output: [1]
```

## 🤔 Understanding "In-Place"

### What does "in-place" mean?

```python
nums1 = [1, 2, 3, 0, 0, 0]  # m = 3 (first 3 are actual elements)
nums2 = [2, 5, 6]            # n = 3

# nums1 has length m + n = 6
# First m=3 elements: [1, 2, 3] are actual values
# Last n=3 elements: [0, 0, 0] are placeholders for nums2
```

**"In-place" means:**
- ❌ You CANNOT create a new array like `result = []`
- ✅ You MUST merge directly into `nums1`
- ✅ Use the extra space already provided in `nums1` (those trailing zeros)
- ✅ Space complexity must be O(1)

## 💡 Exercise 2: Understanding the Right-to-Left Approach

The key insight: **Merge from RIGHT to LEFT** because the extra space is at the END!

**Task**: Study this visualization and understand the pointer movements.

````python
# Visualization of the correct approach
def visualize_merge():
    nums1 = [1, 2, 3, 0, 0, 0]
    nums2 = [2, 5, 6]
    m, n = 3, 3
    
    print("Initial State:")
    print(f"nums1: {nums1}")
    print(f"nums2: {nums2}")
    print(f"       {'  '.join([str(x) for x in range(len(nums1))])}  (indices)")
    print()
    
    i = m - 1      # Last actual element in nums1
    j = n - 1      # Last element in nums2
    k = m + n - 1  # Last position in nums1
    
    print(f"Starting pointers: i={i} (nums1[{i}]={nums1[i]}), j={j} (nums2[{j}]={nums2[j]}), k={k}\n")
    print("="*60)
    
    step = 1
    while i >= 0 and j >= 0:
        print(f"\nStep {step}:")
        print(f"  Compare nums1[{i}]={nums1[i]} vs nums2[{j}]={nums2[j]}")
        
        if nums1[i] > nums2[j]:
            print(f"  → {nums1[i]} > {nums2[j]}, place {nums1[i]} at position {k}")
            nums1[k] = nums1[i]
            i -= 1
        else:
            print(f"  → {nums2[j]} >= {nums1[i]}, place {nums2[j]} at position {k}")
            nums1[k] = nums2[j]
            j -= 1
        k -= 1
        
        print(f"  nums1: {nums1}")
        print(f"  New pointers: i={i}, j={j}, k={k}")
        step += 1
    
    # Copy remaining from nums2 if any
    while j >= 0:
        print(f"\nStep {step}: Copy remaining nums2[{j}]={nums2[j]} to position {k}")
        nums1[k] = nums2[j]
        j -= 1
        k -= 1
        print(f"  nums1: {nums1}")
        step += 1
    
    print("\n" + "="*60)
    print(f"✅ Final Result: {nums1}")

visualize_merge()
````

## ✅ Solution: Complete Implementation

````python
def merge_solution(nums1, m, nums2, n):
    """
    Merge nums2 into nums1 in-place by filling from the end.
    
    Time Complexity: O(m + n) - visit each element once
    Space Complexity: O(1) - no extra space, truly in-place
    """
    # Three pointers
    i = m - 1      # Last actual element in nums1
    j = n - 1      # Last element in nums2
    k = m + n - 1  # Last position in nums1 (where we write)
    
    # Merge from right to left
    while i >= 0 and j >= 0:
        if nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1
    
    # If nums2 has remaining elements, copy them
    # (If nums1 has remaining elements, they're already in correct position)
    while j >= 0:
        nums1[k] = nums2[j]
        j -= 1
        k -= 1


# Alternative: More concise version
def merge_concise(nums1, m, nums2, n):
    """Same logic, more compact"""
    i, j, k = m - 1, n - 1, m + n - 1
    
    while j >= 0:
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1
````

## 🔑 Key Insights

### 1. Why Merge from Right to Left?
- `nums1` has extra space at the **END**
- Merging left-to-right would overwrite data we haven't used yet
- Merging right-to-left uses the empty space first

### 2. Why Only Check `j >= 0` in Second Loop?
```python
while j >= 0:
    nums1[k] = nums2[j]
    j -= 1
    k -= 1
```
- If `nums1` has remaining elements, they're **already in correct position**
- We only need to copy remaining `nums2` elements

### 3. Comparison with Merge Sort

| Aspect | Merge Sort | LeetCode 88 |
|--------|-----------|-------------|
| Direction | Left → Right | Right → Left |
| New Array | Yes (`result = []`) | No (use `nums1`) |
| Space | O(n) | O(1) |
| Use Case | Separate arrays | One array with extra space |

### 4. Edge Cases to Remember
- `m = 0` (nums1 is empty) → Copy all of nums2
- `n = 0` (nums2 is empty) → nums1 already correct
- All `nums2` < all `nums1` → nums2 fills left side
- All `nums2` > all `nums1` → nums1 stays left, nums2 fills right
- Duplicates → Handle with `<=` comparison

### 5. Time & Space Complexity
- **Time**: O(m + n) - each element visited once
- **Space**: O(1) - truly in-place, no extra arrays
