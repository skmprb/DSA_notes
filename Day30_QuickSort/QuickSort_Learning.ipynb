## Summary

### Key Concepts You've Learned:
1. **Partitioning**: Rearranging array around a pivot
2. **Two Pointers**: `i` (boundary) and `j` (scanner)
3. **In-place Sorting**: No extra space needed (unlike merge sort)
4. **Divide and Conquer**: Recursively sort subarrays

### Time Complexity:
- Best Case: O(n log n) - balanced partitions
- Average Case: O(n log n)
- Worst Case: O(n²) - already sorted with last element pivot

### Space Complexity:
- O(log n) - recursion stack space
- O(1) - no extra arrays needed (in-place)

### Quick Sort vs Merge Sort:
| Feature | Quick Sort | Merge Sort |
|---------|-----------|------------|
| Space | O(1) in-place | O(n) extra space |
| Worst case | O(n²) | O(n log n) |
| Stable | No | Yes |
| Cache friendly | Yes | No |
| Practical speed | Faster | Slower |

### When to Use Quick Sort:
- When average case performance matters most
- When space is limited (in-place sorting)
- When stability is not required
- For general-purpose sorting (most languages use it)

### The Magic of `i = low - 1`:
- `i` tracks the boundary: elements at indices `[low...i]` are <= pivot
- Starting at `low - 1` means "no elements <= pivot yet"
- When we find element <= pivot, we increment `i` first, then swap
- Final position `i + 1` is where pivot belongs
