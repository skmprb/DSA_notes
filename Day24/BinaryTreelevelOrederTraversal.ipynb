Given the root of a binary tree, return the level order traversal of its nodes' values. (i.e., from left to right, level by level).

 

Example 1:


Input: root = [3,9,20,null,null,15,7]
Output: [[3],[9,20],[15,7]]
Example 2:

Input: root = [1]
Output: [[1]]
Example 3:

Input: root = []
Output: []
 

Constraints:

The number of nodes in the tree is in the range [0, 2000].
-1000 <= Node.val <= 1000

```python
# Definition for a binary tree node.
from collections import deque
from typing import Optional,List

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        result = []
        queue = deque([root])

        while queue:
            level_size = len(queue)
            level = []

            for _ in range(level_size):
                node = queue.popleft()
                level.append(node.val)

                if node.left:
                    queue.append(node.left)
                if node.right:
                    queue.append(node.right)
            
            result.append(level)
        
        return result

root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(20)

root.left.left = None
root.left.right = None

root.right.left = TreeNode(15)
root.right.right = TreeNode(7)

s = Solution()
print(s.levelOrder(root))


```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        # i: queue to store nodes for current level processing
        i = [root]
        # r: result list containing values grouped by level
        r = [] if root is None else [[root.val]]

        # Continue processing while there are nodes to visit
        while len(i) > 0:
            # t: temporary list to collect child nodes for next level
            t = []
            # r1: list to collect values from current level
            r1 = []
            # Process all nodes in current level
            while len(i) > 0:
                # c: current node being processed
                c = i.pop()   
                if c:             
                    # Add left child to temporary list for next level
                    if c.left:
                        t.insert(0, c.left)
                    # Add right child to temporary list for next level
                    if c.right:
                        t.insert(0, c.right)
            # Move nodes from temporary list to main queue and collect their values
            while len(t) > 0:
                # v: node being moved from temporary list to main queue
                v = t.pop()
                i.insert(0,v)
                r1.append(v.val)
            # Add current level's values to result if any nodes were processed
            if len(r1)>0:
                r.append(r1)
        return r
           
```
