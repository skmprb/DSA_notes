**Key Insight:**
- If match: `dp[i][j] = dp[i-1][j-1]`
- If no match: `dp[i][j] = 1 + min(insert, delete, replace)`

## 🎓 Summary: String DP Patterns

| Pattern | Match | No Match | Use Case |
|---------|-------|----------|----------|
| LCS | 1+dp[i-1][j-1] | max(dp[i-1][j], dp[i][j-1]) | Common subsequence |
| Edit Distance | dp[i-1][j-1] | 1+min(3 ops) | Transform strings |
| LPS | 2+dp[i+1][j-1] | max(dp[i+1][j], dp[i][j-1]) | Palindromes |
| Distinct Subseq | dp[i-1][j-1]+dp[i-1][j] | dp[i-1][j] | Count ways |
| Interleaving | use_s1 OR use_s2 | False | Check formation |

### Common Template:
```python
def string_dp(s1, s2):
    m, n = len(s1), len(s2)
    dp = [[0] * (n+1) for _ in range(m+1)]
    
    # Initialize base cases
    
    for i in range(1, m+1):
        for j in range(1, n+1):
            if s1[i-1] == s2[j-1]:
                dp[i][j] = ...  # Match case
            else:
                dp[i][j] = ...  # No match case
    
    return dp[m][n]
```
