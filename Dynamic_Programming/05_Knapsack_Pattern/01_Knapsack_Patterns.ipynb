## 🎓 Summary: Knapsack Patterns

### 0/1 vs Unbounded

| Type | Iteration | Use Case |
|------|-----------|----------|
| 0/1 | Backward | Each item once |
| Unbounded | Forward | Items unlimited |

### Code Comparison:

**0/1 Knapsack:**
```python
for i in range(n):
    for w in range(capacity, weight[i]-1, -1):  # BACKWARD
        dp[w] = max(dp[w], value[i] + dp[w-weight[i]])
```

**Unbounded Knapsack:**
```python
for i in range(n):
    for w in range(weight[i], capacity+1):  # FORWARD
        dp[w] = max(dp[w], value[i] + dp[w-weight[i]])
```

### Problem Mapping:

| Problem | Type | Conversion |
|---------|------|------------|
| Subset Sum | 0/1 | value = weight |
| Equal Partition | 0/1 | target = total/2 |
| Target Sum | 0/1 | target = (sum+target)/2 |
| Rod Cutting | Unbounded | length = weight |
| Coin Change | Unbounded | coin = weight |
