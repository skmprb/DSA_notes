## 🎓 Summary: Advanced DP Patterns

| Pattern | Key Technique | When to Use |
|---------|---------------|-------------|
| Stock | State tracking | Multiple transactions with constraints |
| Partition DP | Try all splits | Optimal way to partition |
| Tree DP | Return multiple values | Optimization on trees |
| State Machine | Model transitions | Multiple states with rules |
| Bitmask DP | Use bits for sets | Small n, track subsets |

### When to Use Each:

**Stock Problems:**
- Multiple buy/sell with constraints
- Track states: holding, sold, cooldown

**Partition DP:**
- Optimal way to split/group
- Try all partition points
- Usually O(n³)

**Tree DP:**
- Optimization on tree structure
- Return [take, skip] from recursion

**State Machine:**
- Multiple states with transitions
- Paint house, stock with cooldown

**Bitmask DP:**
- n ≤ 20
- Need to track subsets
- Traveling salesman, subset problems
