## 🎓 Key Takeaways

### Memoization Template
```python
def problem(n):
    memo = {}  # or array, or use @lru_cache
    
    def helper(state):
        # 1. Check cache
        if state in memo:
            return memo[state]
        
        # 2. Base case
        if base_condition:
            return base_value
        
        # 3. Compute
        result = recursive_relation(helper(...))
        
        # 4. Store and return
        memo[state] = result
        return result
    
    return helper(n)
```

### When to Use Each Method:
- **Dictionary**: Most flexible, works with any hashable input
- **Array**: Faster for bounded integer inputs
- **@lru_cache**: Cleanest, use when possible!

### Complexity:
- **Time**: O(n) - each state computed once
- **Space**: O(n) for cache + O(n) for recursion stack
