## Part 1: The Problem - Simple Recursion

Let's start with the classic Fibonacci sequence:
- fib(0) = 0
- fib(1) = 1
- fib(n) = fib(n-1) + fib(n-2)

### 🎯 Task 1: Write a Simple Recursive Solution

**Instructions:**
1. Create a function called `fib_recursive` that takes parameter `n`
2. Handle base cases: if n is 0 or 1, return n
3. For other cases, return the sum of fib(n-1) and fib(n-2)

**Hint:** Think about the definition above - translate it directly to code!

```python
# TODO: Write your recursive Fibonacci function here
def fib_recursive(n):
    # Step 1: Handle base cases (when n is 0 or 1)
    # YOUR CODE HERE
    if n <= 1:
        return n
    
    # Step 2: Recursive case - return fib(n-1) + fib(n-2)
    # YOUR CODE HERE
    return fib_recursive(n-1)+fib_recursive(n-2)

# Test your function
print(f"fib(5) = {fib_recursive(5)}")  # Should be 5
print(f"fib(10) = {fib_recursive(10)}")  # Should be 55
```
