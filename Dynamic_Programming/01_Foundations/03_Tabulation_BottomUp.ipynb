## Part 1: Understanding Tabulation

### 📖 Memoization vs Tabulation

**Memoization (Top-Down):**
- Start with the problem
- Break down recursively
- Cache results
- Recursive (uses call stack)

**Tabulation (Bottom-Up):**
- Start with smallest subproblems
- Build up to the answer
- Fill a table iteratively
- Iterative (no recursion!)

### 🎯 Task 1: Fibonacci with Tabulation

**Your Mission:** Build the Fibonacci sequence from bottom-up

**Think:**
1. Create a table (array) to store results
2. Fill in base cases (fib(0), fib(1))
3. Use a loop to fill remaining values
4. Each value uses previously computed values

````python
def fib_tabulation(n):
    """
    Calculate Fibonacci using tabulation (bottom-up).
    
    Steps:
    1. Handle edge case (n <= 1)
    2. Create DP table of size n+1
    3. Initialize base cases: dp[0]=0, dp[1]=1
    4. Loop from 2 to n, filling dp[i] = dp[i-1] + dp[i-2]
    5. Return dp[n]
    """
    
    # TODO: Step 1 - Handle edge case
    # Hint: if n <= 1:
    #           return n
    # YOUR CODE HERE

    if n <= 1:
        return n
    
    # TODO: Step 2 - Create DP table
    # Hint: dp = [0] * (n + 1)
    # YOUR CODE HERE

    dp = [0] * (n+1)
    
    # TODO: Step 3 - Initialize base cases
    # Hint: dp[0] = 0
    #       dp[1] = 1
    # YOUR CODE HERE

    dp[0] = 0
    dp[1] = 1
    
    # TODO: Step 4 - Fill the table
    # Hint: for i in range(2, n + 1):
    #           dp[i] = dp[i-1] + dp[i-2]
    # YOUR CODE HERE

    for i in range(2,n+1):
        dp[i] = dp[i-1] + dp[i-2]
    
    # TODO: Step 5 - Return final answer
    # Hint: return dp[n]
    # YOUR CODE HERE
    return dp[n]
    

# Test your implementation
print(f"fib(10) = {fib_tabulation(10)}")
print(f"fib(50) = {fib_tabulation(50)}")
print(f"fib(100) = {fib_tabulation(100)}")
````

````python
# ✅ Verify your solution
assert fib_tabulation(10) == 55, "fib(10) should be 55"
assert fib_tabulation(20) == 6765, "fib(20) should be 6765"
print("✅ Excellent! Tabulation works!")
````

## Part 2: Space Optimization

### 📖 Key Observation
To calculate fib(n), we only need fib(n-1) and fib(n-2).
We don't need the entire array!

**Space: O(n) → O(1)**

### 🎯 Task 3: Optimize Space

**Your Mission:** Use only 2 variables instead of an array

````python
def fib_optimized(n):
    """
    Calculate Fibonacci with O(1) space.
    
    Strategy:
    - Keep only last 2 values: prev2 and prev1
    - Calculate current = prev1 + prev2
    - Shift: prev2 = prev1, prev1 = current
    """
    
    # TODO: Handle edge case
    # YOUR CODE HERE
    if n <= 1:
        return n
    
    # TODO: Initialize prev2 and prev1
    # Hint: prev2 = 0  # fib(0)
    #       prev1 = 1  # fib(1)
    # YOUR CODE HERE
    prev2 = 0
    prev1 = 1
    
    # TODO: Loop from 2 to n
    # Calculate current, then shift values
    # Hint: for i in range(2, n + 1):
    #           curr = prev1 + prev2
    #           prev2 = prev1
    #           prev1 = curr
    # YOUR CODE HERE

    for i in range(2,n+1):
        prev1, prev2 = prev1+prev2 , prev1
    
    # TODO: Return prev1
    # YOUR CODE HERE
    return prev1
    pass

# Test your implementation
print(f"fib(10) = {fib_optimized(10)}")
print(f"fib(100) = {fib_optimized(100)}")
````

````python
# ✅ Verify your solution
assert fib_optimized(10) == 55, "fib(10) should be 55"
assert fib_optimized(50) == 12586269025, "fib(50) should be 12586269025"
print("✅ Amazing! You've optimized the space!")
````

## Part 3: Real Problem - Climbing Stairs

### 📖 Problem
Count ways to climb n stairs (1 or 2 steps at a time).

### 🎯 Task 4: Implement with Tabulation

````python
def climb_stairs_tabulation(n):
    """
    Count ways to climb n stairs using tabulation.
    
    dp[i] = number of ways to reach step i
    dp[i] = dp[i-1] + dp[i-2]
    """
    # TODO: Implement using tabulation approach
    # Follow same pattern as Fibonacci!
    # YOUR CODE HERE
    pass

    if n <= 1:
        return 1
    dp = [0] * (n+1)

    dp[0] = dp[1] = 1

    for i in range(2,n+1):
        dp[i] = dp[i-1] + dp[i-2]
    
    return dp[n]

# Test
print(f"Ways to climb 5 stairs: {climb_stairs_tabulation(5)}")
print(f"Ways to climb 10 stairs: {climb_stairs_tabulation(10)}")
````

````python
# Now optimize the space!
def climb_stairs_optimized(n):
    """
    Space-optimized version using O(1) space.
    """
    # TODO: Use only 2 variables
    # YOUR CODE HERE
    pass

    if n <= 1:
        return 1
    
    prev2 = 1
    prev1 = 1

    for i in range(2, n+1):
        prev2, prev1 = prev1, prev1+prev2
    
    print("done")
    return prev1

print(f"Ways to climb 1000 stairs: {climb_stairs_optimized(1000)}")
````

## Part 4: Real Problem - Min Cost Climbing Stairs

### 📖 Problem
Given cost array, find minimum cost to reach top.
Can start from index 0 or 1.

**Example:** cost = [10, 15, 20] → Output: 15

### 🎯 Task 5: Implement with Tabulation

**Think:**
- dp[i] = minimum cost to reach step i
- To reach step i, come from i-1 or i-2
- dp[i] = min(dp[i-1] + cost[i-1], dp[i-2] + cost[i-2])

````python
def min_cost_climbing_tabulation(cost):
    """
    Find minimum cost to reach top using tabulation.
    
    Steps:
    1. Create dp array of size len(cost) + 1
    2. dp[0] = 0, dp[1] = 0 (can start from either)
    3. For i from 2 to n:
       dp[i] = min(dp[i-1] + cost[i-1], dp[i-2] + cost[i-2])
    4. Return dp[n]
    """
    n = len(cost)
    
    # TODO: Create DP table
    # YOUR CODE HERE

    dp = [0]*(n+1)
    
    # TODO: Initialize base cases
    # YOUR CODE HERE

    if n <= 1:
        return 1
    
    # TODO: Fill the table
    # YOUR CODE HERE

    for i in range(2,n+1):
        dp[i] = min(dp[i-1] + cost[i-1], dp[i-2]+cost[i-2])
    
    # TODO: Return answer
    # YOUR CODE HERE
    return dp[n]
    pass

# Test
cost1 = [10, 15, 20]
print(f"Min cost: {min_cost_climbing_tabulation(cost1)}")  # Should be 15

cost2 = [1, 100, 1, 1, 1, 100, 1, 1, 100, 1]
print(f"Min cost: {min_cost_climbing_tabulation(cost2)}")  # Should be 6
````

````python
# Now optimize space to O(1)!
def min_cost_climbing_optimized(cost):
    """
    Space-optimized version.
    Only need last 2 values!
    """
    # TODO: Use only 2 variables
    # YOUR CODE HERE
    n = len(cost)
    if n <= 1:
        return 1
    
    prev2 = 0
    prev1 = 0

    for i in range(2, n+1):
        prev2, prev1 = prev1, min(prev1 + cost[i-1], prev2 + cost[i-2])
    return prev1

cost2 = [1, 100, 1, 1, 1, 100, 1, 1, 100, 1]
print(f"Min cost (optimized): {min_cost_climbing_optimized(cost2)}")
````

````python
# ✅ Verify
assert min_cost_climbing_tabulation([10, 15, 20]) == 15, "Should be 15"
assert min_cost_climbing_optimized([1, 100, 1, 1, 1, 100, 1, 1, 100, 1]) == 6, "Should be 6"
print("✅ Brilliant! You're a tabulation expert!")
````

## Part 5: Real Problem - House Robber

### 📖 Problem
Rob houses along a street. Can't rob adjacent houses.
Find maximum money you can rob.

**Example:** [2, 7, 9, 3, 1] → Output: 12 (rob 2, 9, 1)

**Analysis:**
- For each house, decide: rob it or skip it
- If rob house i: money[i] + dp[i-2]
- If skip house i: dp[i-1]
- dp[i] = max(rob, skip)

### 🎯 Task 6: Implement House Robber

````python
def house_robber(nums):
    """
    Find maximum money that can be robbed.
    
    dp[i] = max money robbing houses 0 to i
    dp[i] = max(nums[i] + dp[i-2], dp[i-1])
    """
    # TODO: Handle edge cases
    # YOUR CODE HERE

    if not nums:
        return 0
    
    if len(nums) == 1:
        return nums[0]
    
    n = len(nums)
    # TODO: Create DP table
    # YOUR CODE HERE

    dp = [0]*(n)
    
    # TODO: Initialize base cases
    # dp[0] = nums[0]
    # dp[1] = max(nums[0], nums[1])
    # YOUR CODE HERE

    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])
    
    # TODO: Fill table
    # For each house, choose to rob or skip
    # YOUR CODE HERE

    for i in range(2,n):
        dp[i] = max(nums[i] + dp[i-2], dp[i-1])
    
    # TODO: Return answer
    # YOUR CODE HERE
    return dp[n-1]

# Test
houses = [2, 7, 9, 3, 1]
print(f"Max money: {house_robber(houses)}")  # Should be 12
````

````python
# Space-optimized version
def house_robber_optimized(nums):
    """
    O(1) space solution.
    """
    # TODO: Use only 2 variables
    # YOUR CODE HERE
    n = len(nums)

    prev2 = nums[0]
    prev1 = max(nums[0], nums[1])

    for i in range(2,n):
        prev2, prev1 = prev1, max(nums[i]+prev2 , prev1)
    
    return prev1

print(f"Max money (optimized): {house_robber_optimized([2, 7, 9, 3, 1])}")
````

## 🎓 Key Takeaways

### Tabulation Template
```python
def problem(n):
    # 1. Create DP table
    dp = [0] * (n + 1)
    
    # 2. Initialize base cases
    dp[0] = base_value_0
    dp[1] = base_value_1
    
    # 3. Fill table using recurrence
    for i in range(2, n + 1):
        dp[i] = f(dp[i-1], dp[i-2], ...)
    
    # 4. Return answer
    return dp[n]
```

### Space Optimization Template
```python
def problem_optimized(n):
    # Keep only last k values
    prev2 = base_value_0
    prev1 = base_value_1
    
    for i in range(2, n + 1):
        curr = f(prev1, prev2)
        prev2 = prev1
        prev1 = curr
    
    return prev1
```

### Memoization vs Tabulation

| Aspect | Memoization | Tabulation |
|--------|-------------|------------|
| Approach | Top-Down | Bottom-Up |
| Style | Recursive | Iterative |
| Space | O(n) + stack | O(n) or O(1) |
| When | Natural recursion | Clear iteration |
| Pros | Intuitive | Faster, no stack |
