## 🎓 Key Takeaways

### Common DP Patterns

| Pattern | State | Example |
|---------|-------|----------|
| Linear (1D) | dp[i] | Fibonacci, Climbing Stairs |
| Grid (2D) | dp[i][j] | Unique Paths, Min Path Sum |
| String (2D) | dp[i][j] | LCS, Edit Distance |
| Knapsack | dp[i][w] | 0/1 Knapsack, Subset Sum |

### Keywords Cheat Sheet

**See "maximum/minimum"** → Likely DP  
**See "count ways"** → Likely DP  
**See "longest/shortest"** → Likely DP  
**See "is it possible"** → Maybe DP  

### The Framework (Memorize This!)

1. **Identify** - Is it DP?
2. **State** - What changes?
3. **Recurrence** - How to compute current from previous?
4. **Base Cases** - Smallest subproblems?
5. **Approach** - Memo or Tab?
6. **Optimize** - Can reduce space?
