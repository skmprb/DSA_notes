## 🎓 Summary: Grid DP Patterns

| Pattern | Recurrence | Space Opt |
|---------|------------|----------|
| Path Counting | dp[i][j] = dp[i-1][j] + dp[i][j-1] | O(n) |
| Min Path Sum | dp[i][j] = grid[i][j] + min(...) | O(n) |
| Triangle | dp[i][j] = triangle[i][j] + min(...) | O(n) |
| Maximal Square | dp[i][j] = min(3 neighbors) + 1 | O(n) |

### Common Template:
```python
def grid_dp(grid):
    m, n = len(grid), len(grid[0])
    dp = [[0] * n for _ in range(m)]
    
    # Initialize first row and column
    dp[0][0] = grid[0][0]
    
    # Fill table
    for i in range(m):
        for j in range(n):
            dp[i][j] = f(dp[i-1][j], dp[i][j-1])
    
    return dp[m-1][n-1]
```
