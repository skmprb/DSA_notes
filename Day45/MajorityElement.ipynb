169. Majority Element

Given an array nums of size n, return the majority element.

The majority element is the element that appears more than ⌊n / 2⌋ times. You may assume that the majority element always exists in the array.

 

Example 1:

Input: nums = [3,2,3]
Output: 3
Example 2:

Input: nums = [2,2,1,1,1,2,2]
Output: 2
 

Constraints:

n == nums.length
1 <= n <= 5 * 104
-109 <= nums[i] <= 109
The input is generated such that a majority element will exist in the array.
 

Follow-up: Could you solve the problem in linear time and in O(1) space?

```python
from collections import Counter
from typing import List
nums = [3,2,3]
count = Counter(nums)
half = len(nums)/2
for i in range(len(nums)):
    if count.get(nums[i]) >= half:
        print(nums[i])
        break
```

```python
class Solution:
    def majorityElement(self, nums: List[int]) -> int:
        count = Counter(nums)
        half = len(nums)/2
        for i in range(len(nums)):
            if count.get(nums[i]) >= half:
                return nums[i]
        
```

```python
class Solution:
    def majorityElement(self, nums: List[int]) -> int:
        nums.sort()
        return nums[len(nums)//2]
```

```python
class Solution:
    def majorityElement(self, nums: List[int]) -> int:
        num,count=nums[0],1
        print(f"Initial: num={num}, count={count}")
        for i in range(1,len(nums)):
            print(f"Processing nums[{i}]={nums[i]}")
            if nums[i]==num:
                count+=1
                print(f"  Same as current num: count incremented to {count}")
            else:
                count-=1
                print(f"  Different from current num: count decremented to {count}")
            if count<0:
                num=nums[i]
                count=1
                print(f"  Count went negative, switching: num={num}, count={count}")
            print(f"  Current state: num={num}, count={count}")
        print(f"Final result: {num}")
        return num
s = Solution()
print(s.majorityElement([2,2,1,1,1,2,2]))
```
