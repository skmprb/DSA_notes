16. 3Sum Closest

Given an integer array nums of length n and an integer target, find three integers at distinct indices in nums such that the sum is closest to target.

Return the sum of the three integers.

You may assume that each input would have exactly one solution.

Example 1:

Input: nums = [-1,2,1,-4], target = 1
Output: 2
Explanation: The sum that is closest to the target is 2. (-1 + 2 + 1 = 2).
Example 2:

Input: nums = [0,0,0], target = 1
Output: 0
Explanation: The sum that is closest to the target is 0. (0 + 0 + 0 = 0).
 

Constraints:

3 <= nums.length <= 500
-1000 <= nums[i] <= 1000
-104 <= target <= 104

## problem solution

sort the array
one for loop for the index i
while with ponter of left and right

in while we check 
the check the sum of i + left + right and if its same as target return it,
if the sum - target less then closest - sum then make the sum as closest.

after check return the closest

```python
def closeSum(nums,target):
    n = len(nums)
    nums.sort()
    closest = float("inf")
    for i in range(n):
        left = i+1
        right = n-1
        while left < right:
            current_sum = nums[i] + nums[left] + nums[right]
            if current_sum == target:
                return current_sum
            
            if abs(current_sum - target) < abs(closest - target):
                closest = current_sum
            
            if current_sum < target:
                left += 1
            else:
                right -= 1
    return closest

print(closeSum([-1,2,1,-4], 1))
print(closeSum([0,0,0], 1))           # Expected: 0
print(closeSum([1,1,1,0], -100))      # Expected: 2
print(closeSum([0,2,1,-3], 1))        # Expected: 0          
```

```python
from typing import List

class Solution:
    def threeSumClosest(self, nums: List[int], target: int) -> int:
        closest = float("inf")
        nums.sort()
        for i in range(len(nums)):

            if i > 0 and nums[i] == nums[i - 1]: continue 

            sum = nums[i] + nums[i + 1] + nums[i + 2]
            if sum > target:
                if sum - target < abs(closest - target):
                    closest = sum
                break


            sum = nums[i] + nums[-2] + nums[-1]
            if sum < target: 
                if target - sum < abs(closest - target):
                    closest = sum
                continue 

            left = i+1
            right = len(nums) - 1
            while left < right:
                current_sum = nums[i] + nums[left] + nums[right]
                if current_sum == target:
                    return current_sum
                if abs(current_sum - target) < abs(closest - target):
                    closest = current_sum
                if current_sum < target:
                    left += 1
                else:
                    right -= 1
        return closest

s= Solution()
print(s.threeSumClosest([-1,2,1,-4], 1))  
print(s.threeSumClosest([0,0,0], 1))  
print(s.threeSumClosest([1,1,1,0], -100))  
print(s.threeSumClosest([0,2,1,-3], 1))
```

```python
class Solution:
    def threeSumClosest(self, nums: List[int], target: int) -> int:
        # This function finds three numbers in the array whose sum is closest to the target value
        # Time O(n^2), sorting O(nlogn), nested loop O(n^2) # Space O(1), ignoring sorting stack overhead
        nums.sort()
        n = len(nums)
        # Initialize answer to positive infinity to ensure any valid sum will be closer
        ans = float("inf") # positive infinity, inf is float type
        # Iterate through array, leaving space for two more elements (j and k)
        for i in range(n-2): # leave space for j and k
            # Skip duplicate values to avoid redundant calculations
            # check if i is duplicate
            if i > 0 and nums[i] == nums[i - 1]: continue # this number is the same as the previous number
            # Optimization 1: Check minimum possible sum with current i
            # If the smallest possible sum (i + next two elements) is already > target
            # optimization 1
            sum = nums[i] + nums[i + 1] + nums[i + 2]
            if sum > target: # any three numbers after this will have sum greater than current sum
                if sum - target < abs(ans - target):
                    ans = sum
                break # early termination, follows the first if
            # Optimization 2: Check maximum possible sum with current i
            # If the largest possible sum (i + last two elements) is < target
            # optimization 2
            sum = nums[i] + nums[-2] + nums[-1]
            if sum < target: # nums[i] plus any two numbers after it won't exceed sum, so the two-pointer below doesn't need to run, nums[i] is too small
                if target - sum < abs(ans - target):
                    ans = sum
                continue # skip this nums[i], follows the first if

            # Two-pointer technique to find the closest sum
            j = i + 1
            k = n - 1
            while j < k:
                sum = nums[i] + nums[j] + nums[k]
                # Update answer if current sum is closer to target
                if abs(sum - target) < abs(ans - target):
                    ans = sum
                # If exact match found, return immediately
                if sum == target:
                    return target
                # Move pointers based on whether sum is greater or less than target
                elif sum > target:
                    k -= 1
                else:
                    j += 1
        return ans
                
s= Solution()
print(s.threeSumClosest([-1,2,1,-4], 1)) 
```
