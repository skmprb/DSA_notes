168. Excel Sheet Column Title

Given an integer columnNumber, return its corresponding column title as it appears in an Excel sheet.

For example:

A -> 1
B -> 2
C -> 3
...
Z -> 26
AA -> 27
AB -> 28 
...
 

Example 1:

Input: columnNumber = 1
Output: "A"
Example 2:

Input: columnNumber = 28
Output: "AB"
Example 3:

Input: columnNumber = 701
Output: "ZY"
 

Constraints:

1 <= columnNumber <= 231 - 1

## Problem state

here the can be as for 1 to 26 it would be A to Z,
if number is more than 26 then we take reminder ex: 28 // 26 which is 2 and 
since it is less then 26 * 2 the start carrector  is A and for the 2 we will add B, foloowing like this

```python
def converttotitle(columnNumber):

    # Fixed: Removed duplicate print statement
    print(f"Starting with columnNumber: {columnNumber}")
    result = ""
    # Fixed: Moved this print inside the loop where it belongs
    while columnNumber > 0:
        print(f"Loop iteration - columnNumber: {columnNumber}")
        # Convert to 0-indexed
        columnNumber -= 1
        print(f"After converting to 0-indexed: {columnNumber}")
        # Fixed: Moved remainder calculation before its print statement
        remainder = columnNumber % 26
        print(f"Remainder (columnNumber % 26): {remainder}")

        # Fixed: Moved char calculation before its print statement
        char = chr(ord('A') + remainder)
        print(f"Character: {char}")
        # Get the rightmost character
        result = char + result  # Prepend to result
        print(f"Result so far: {result}")
        # Move to next digit
        columnNumber = columnNumber // 26
        print(f"Next columnNumber (// 26): {columnNumber}")
        print("---")

    print(f"Final result: {result}")
    return result

print(converttotitle(701))

```

```python
class Solution:
    def convertToTitle(self, columnNumber: int) -> str:
        result = ""
        while columnNumber > 0:
            columnNumber -= 1

            remainder = columnNumber % 26
            char = chr(ord('A') + remainder)
            result = char + result

            columnNumber = columnNumber // 26

        return result
    
s = Solution()
print(s.convertToTitle(701))
```

```python
class Solution:
    def convertToTitle(self, columnNumber: int) -> str:
        result = ""
        while columnNumber > 0:
            columnNumber -= 1
            result = chr(columnNumber % 26 + ord('A')) + result
            columnNumber //= 26
        return result
```
