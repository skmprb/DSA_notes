```python
# This sorting algorithm is called Selection Sort
# It works by repeatedly finding the minimum element from the unsorted portion
# and placing it at the beginning of the sorted portion

# def sorting(nums):
#     for i in range(len(nums)):
#         for j in range(i,len(nums)):
#             if nums[i] > nums[j]:
#                 nums[i], nums[j] = nums[j], nums[i]
#     return nums

# print(sorting([5,3,2,1,4]))

def sorting(nums):
    for i in range(len(nums)-1):
        minindex = i
        for j in range(i+1, len(nums)):
            if nums[j] < nums[minindex]:
                minindex = j
        nums[i], nums[minindex] = nums[minindex], nums[i]
    return nums

print(sorting([5, 3, 2, 1, 4]))
```

```python
## Merge Sort

def merge(nums, left,mid,right):
    # Print the merge operation being performed
    print(f"Merging: left={left}, mid={mid}, right={right}")
    print(f"Left subarray: {nums[left:mid+1]}, Right subarray: {nums[mid+1:right+1]}")
    
    n1 = mid-left +1
    n2 = right - mid

    Leftnum = [0]*n1
    Rightnum = [0]*n2

    for i in range(n1):
        Leftnum[i] = nums[left+i]
    for j in range(n2):
        Rightnum[j] = nums[mid+1+j]
    
    print(f"Left temp array: {Leftnum}")
    print(f"Right temp array: {Rightnum}")
    
    i = j = 0
    k = left

    while i < n1 and j < n2:
        if Leftnum[i] <= Rightnum[j]:
            nums[k] = Leftnum[i]
            print(f"Choosing {Leftnum[i]} from left array")
            i+=1
        else:
            nums[k] = Rightnum[j]
            print(f"Choosing {Rightnum[j]} from right array")
            j+=1
        
        k+=1
    
    while i < n1:
        nums[k] = Leftnum[i]
        print(f"Adding remaining {Leftnum[i]} from left array")
        i+=1
        k+=1
    
    while j < n2:
        nums[k] = Rightnum[j]
        print(f"Adding remaining {Rightnum[j]} from right array")
        j+=1
        k+=1
    
    print(f"After merge: {nums[left:right+1]}")
    print("-" * 40)
    return nums



def mergesort(nums, left, right):
    # Bug: These parameters should not be overridden
    # Print the current recursive call
    print(f"MergeSort called with: array={nums[left:right+1]}, left={left}, right={right}")

    if left < right:
        mid = left+ (right-left)//2
        print(f"Dividing at mid={mid}")
        mergesort(nums, left, mid)
        mergesort(nums, mid+1,right)
        merge(nums,left,mid,right)
    else:
        print(f"Base case reached: single element {nums[left] if left <= right else 'empty'}")
    return nums


arr = [12, 11, 13, 5, 6, 7]
print("Initial array:", arr)
print("=" * 50)
result = mergesort(arr,0, len(arr)-1)
print("=" * 50)
print("Final sorted array:", result)
```

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


print(merge_sort([8, 3, 5, 4, 7, 6, 1, 2]))
```
