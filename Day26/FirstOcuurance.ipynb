Given two strings needle and haystack, return the index of the first occurrence of needle in haystack, or -1 if needle is not part of haystack.

 

Example 1:

Input: haystack = "sadbutsad", needle = "sad"
Output: 0
Explanation: "sad" occurs at index 0 and 6.
The first occurrence is at index 0, so we return 0.
Example 2:

Input: haystack = "leetcode", needle = "leeto"
Output: -1
Explanation: "leeto" did not occur in "leetcode", so we return -1.
 

Constraints:

1 <= haystack.length, needle.length <= 104
haystack and needle consist of only lowercase English characters.

```python
# Time Complexity: O(n*m) where n is length of haystack and m is length of needle
# Space Complexity: O(1) - only using constant extra space
def strStr(haystack,needle):
    for i in range(len(haystack)):
        if needle[0] == haystack[i]:
            start = i
            j = 1
            while j < len(needle):
                i+=1
                if needle[j] != haystack[i]:
                    return -1
                j+=1
            return start

print(strStr("sadbutsad","sad"))

# Time Complexity: O(n*m) where n is length of haystack and m is length of needle (worst case for 'in' operator)
# Space Complexity: O(1) - only using constant extra space
def strStr(haystack, needle):
    if needle not in haystack:
        return -1
    else:
        return haystack.index(needle)

# Time Complexity: O(n*m) where n is length of haystack and m is length of needle
# Space Complexity: O(1) - only using constant extra space
def strStr2(haystack, needle):
    return haystack.find(needle)

# Time Complexity: O(n*m) where n is length of haystack and m is length of needle
# Space Complexity: O(m) - creating substring of length m for comparison
def strStr3(haystack, needle):
    for i in range(len(haystack) - len(needle) + 1):
        if haystack[i:i+len(needle)] == needle:
            return i
    return -1


# Time Complexity: O(n*m) where n is length of haystack and m is length of needle
# Space Complexity: O(1) - only using constant extra space
def strStr4(haystack, needle):
    for i in range(len(haystack) - len(needle) + 1):
        match = True
        for j in range(len(needle)):
            if haystack[i+j] != needle[j]:
                match = False
                break
        if match:
            return i
    return -1

print(strStr3("sadbutsad", "sad"))

# Time Complexity: O(n*m) where n is length of haystack and m is length of needle
# Space Complexity: O(m) - creating substring of length m for comparison
def strStr5(haystack,needle):
    for i in range(len(haystack)):
        if needle[0] == haystack[i]:
            if haystack[i:i+len(needle)] == needle:
                return i
            else:
                return -1

```
