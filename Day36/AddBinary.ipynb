67. Add Binary

Given two binary strings a and b, return their sum as a binary string.

 

Example 1:

Input: a = "11", b = "1"
Output: "100"
Example 2:

Input: a = "1010", b = "1011"
Output: "10101"
 

Constraints:

1 <= a.length, b.length <= 104
a and b consist only of '0' or '1' characters.
Each string does not contain leading zeros except for the zero itself.

```python
# This class implements binary addition for two binary strings
class Solution:
    def addBinary(self, a: str, b: str) -> str:
        # Initialize result list to store binary digits and carry for overflow
        result = []
        carry = 0
        # Start from the rightmost digits of both strings
        i, j = len(a) - 1, len(b) - 1
        
        # Continue while there are digits to process or carry exists
        while i >= 0 or j >= 0 or carry:
            # Start with carry from previous addition
            total = carry
            
            # Add digit from string 'a' if available
            if i >= 0:
                total += int(a[i])
                i -= 1
            
            # Add digit from string 'b' if available
            if j >= 0:
                total += int(b[j])
                j -= 1
            
            # Append the result digit (0 or 1) to result list
            result.append(str(total % 2))
            # Calculate carry for next position (0 or 1)
            carry = total // 2
        
        # Reverse the result since we built it backwards and join into string
        return ''.join(reversed(result))
        
# Create instance and test with binary strings "11" (3 in decimal) and "1" (1 in decimal)
# Expected result: "100" (4 in decimal)
s = Solution()
print(s.addBinary("11","1"))
```

```python
class Solution:
    def addBinary(self, a: str, b: str) -> str:
        # Convert binary string 'a' to decimal integer using base 2
        c=int(a,2)
        # Convert binary string 'b' to decimal integer using base 2
        d=int(b,2)
        # Add the two decimal numbers together
        e=c+d
        # Convert the sum back to binary string and remove the '0b' prefix
        return bin(e)[2:]

# Create an instance of the Solution class
s = Solution()
# Test the addBinary method with "11" (binary 3) and "1" (binary 1)
# Expected result: "100" (binary 4, which is 3+1)
print(s.addBinary("11","1"))
```
