215. Kth Largest Element in an Array
Given an integer array nums and an integer k, return the kth largest element in the array.

Note that it is the kth largest element in the sorted order, not the kth distinct element.

Can you solve it without sorting?

 

Example 1:

Input: nums = [3,2,1,5,6,4], k = 2
Output: 5
Example 2:

Input: nums = [3,2,3,1,2,4,5,5,6], k = 4
Output: 4
 

Constraints:

1 <= k <= nums.length <= 105
-104 <= nums[i] <= 104

```python
from typing import List

class Solution:

    def findKthLargest(self, nums: List[int], k: int) -> int:
        k = len(nums) - k
        def partition(low, high):
            pivot = nums[high]
            i = low-1
            for j in range(low,high):
                if nums[j] <= pivot:
                    i+=1
                    nums[i], nums[j] = nums[j], nums[i]
            
            nums[i+1], nums[high] = nums[high], nums[i+1]
            return i+1
        
        def quickselect(low, high):
            p = partition(low,high)
            if p == k:
                return nums[p]
            elif p < k:
                return quickselect(p+1,high)
            else:
                return quickselect(low,p-1)

        return quickselect(0, len(nums)-1)
    
s = Solution()
print(s.findKthLargest([3,2,1,5,6,4], 2))
            
```

```python
from typing import List

class Solution:

    def findKthLargest(self, nums: List[int], k: int) -> int:
        k = len(nums) - k
        def three_way_partition(low, high):
            if low >= high:
                return low, high
            pivot = nums[high]
            j = low
            i = low
            p = high
            
            while j < p:
                if nums[j] < pivot:
                    nums[i], nums[j] = nums[j], nums[i]
                    i+=1
                    j+=1
                elif nums[j] > pivot:
                    p -= 1
                    nums[j], nums[p] = nums[p], nums[j]
                else:
                    j+=1
            
            nums[p], nums[high] = nums[high], nums[p]
            return i, p
        
        def quickselect(low, high):
            if low <= high:
                left,right = three_way_partition(low,high)
                if k < left:
                    return quickselect(low, left-1)
                elif k > right:
                    return quickselect(right-1, high)
                else:
                    return nums[k]

        return quickselect(0, len(nums)-1)
    
s = Solution()
print(s.findKthLargest([2,1], 1))
            
```

```python
import heapq

# This function finds the Kth largest element in an array using a min-heap approach
# It maintains a heap of size k containing the k largest elements seen so far
# The root of this min-heap will always be the kth largest element
def findKthLargest(nums,k):
    heap = []

    # Iterate through each number in the input array
    for num in nums:
        # Push the current number onto the heap
        heapq.heappush(heap,num)
        # If heap size exceeds k, remove the smallest element
        # This ensures we only keep the k largest elements
        if len(heap) > k:
            heapq.heappop(heap)
    # Return the smallest element in the heap, which is the kth largest overall
    return heap[0]

# heapq is Python's built-in module for heap queue algorithm (priority queue)
# It implements a binary min-heap where the smallest element is at index 0
# heappush(heap, item) - pushes item onto heap, maintaining heap invariant
# heappop(heap) - pops and returns smallest item from heap
# Time complexity: O(n log k) where n is array length and k is the target position
# Space complexity: O(k) for storing k elements in the heap

print(findKthLargest([2,1],1))
```

```python
class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        # Manual min heap implementation
        heap = []
        
        def parent(i):
            return (i - 1) // 2
        
        def left_child(i):
            return 2 * i + 1
        
        def right_child(i):
            return 2 * i + 2
        
        def heapify_up(i):
            # Bubble up to maintain min heap property
            while i > 0 and heap[parent(i)] > heap[i]:
                heap[parent(i)], heap[i] = heap[i], heap[parent(i)]
                i = parent(i)
        
        def heapify_down(i):
            # Bubble down to maintain min heap property
            min_idx = i
            left = left_child(i)
            right = right_child(i)
            
            if left < len(heap) and heap[left] < heap[min_idx]:
                min_idx = left
            if right < len(heap) and heap[right] < heap[min_idx]:
                min_idx = right
            
            if min_idx != i:
                heap[i], heap[min_idx] = heap[min_idx], heap[i]
                heapify_down(min_idx)
        
        def push(val):
            heap.append(val)
            heapify_up(len(heap) - 1)
        
        def pop():
            if len(heap) == 0:
                return None
            if len(heap) == 1:
                return heap.pop()
            
            root = heap[0]
            heap[0] = heap.pop()
            heapify_down(0)
            return root
        
        # Build heap of size k
        for num in nums:
            push(num)
            if len(heap) > k:
                pop()
        
        return heap[0]
```
