## Part 1: Understanding Heap Structure

### Concept: Array as Binary Tree
```
Array: [50, 30, 20, 15, 10, 8, 16]
Index:  0   1   2   3   4   5   6

Tree representation:
         50 (i=0)
        /  \
      30    20
     /  \  /  \
    15  10 8  16
```

**Key Formulas:**
- Parent of index `i` = `(i-1) // 2`
- Left child of index `i` = `2*i + 1`
- Right child of index `i` = `2*i + 2`

## Part 2: Understanding Max Heap Property

### Concept: Max Heap
**Max Heap Property:** Every parent >= its children

```
Valid Max Heap:        Invalid (not a heap):
      50                     20
     /  \                   /  \
    30  20                 50  10
   /  \                   /
  15  10                 30
```

### Exercise 2.1: Check if Node Satisfies Max Heap Property
Given array and index `i`, check if node at `i` is >= its children.

````python
def is_max_heap_at(arr, n, i):
    # TODO: Check if arr[i] >= its children
    # Remember: left = 2*i+1, right = 2*i+2
    # Check if children exist (< n) before comparing
    
    left = ((2 * i ) + 1)
    right = ((2 * i) + 2)
    if left < n and arr[i] < arr[left]:
        return False
    if right < n and arr[i] < arr[right]:
        return False
    
    
    return True


# Test
arr1 = [50, 30, 20, 15, 10]
print(is_max_heap_at(arr1, 5, 0))  # True (50 >= 30 and 50 >= 20)
print(is_max_heap_at(arr1, 5, 1))  # True (30 >= 15 and 30 >= 10)

arr2 = [20, 50, 10, 30]
print(is_max_heap_at(arr2, 4, 0))  # False (20 < 50)
````

## Part 3: Heapify - Fixing Heap Property

### Concept: Heapify
**Heapify** fixes heap property violation at index `i` by:
1. Finding largest among parent and children
2. If largest is not parent, swap and recursively heapify the affected child

```
Example: Heapify at index 0
Before:        After swap:      After recursive heapify:
   10             50                  50
  /  \           /  \                /  \
 50  20         10  20              30  20
 /              /                   /
30             30                  10
```

### Exercise 3.1: Heapify Without Recursion (Single Level)
Fix heap property at index `i` by swapping with largest child (no recursion yet).

````python
def heapify_single(arr, n, i):
    # TODO: Find largest among parent and children
    # If largest != i, swap arr[i] with arr[largest]
    largest = i
    left = ((2*i)+1)
    right = ((2*i)+2)
    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right
    
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]



# Test
arr = [10, 50, 20, 30]
heapify_single(arr, 4, 0)
print(arr)  # Should be [50, 10, 20, 30] or [50, 30, 20, 10]
````

## Part 4: Build Max Heap

### Concept: Why Start from `n//2 - 1`?
**Leaf nodes** (indices `n//2` to `n-1`) are already valid heaps (no children).
**Internal nodes** (indices `0` to `n//2-1`) need heapify.

```
Array: [4, 10, 3, 5, 1]
Index:  0   1  2  3  4

Tree:      4
          / \
        10   3
        / \
       5   1

n = 5, n//2 = 2
Leaf nodes: indices 2, 3, 4 (no children)
Internal nodes: indices 0, 1 (have children)
Start heapify from index n//2-1 = 1, then 0
```

## Part 5: HeapSort Algorithm

### Concept: HeapSort Process
1. **Build max heap** - Largest element is at index 0
2. **Extract max repeatedly:**
   - Swap arr[0] with arr[n-1] (move max to end)
   - Reduce heap size by 1
   - Heapify root (index 0) to restore heap property
   - Repeat until heap size = 1

```
Example: [4, 10, 3, 5, 1]

Step 1: Build heap → [10, 5, 3, 4, 1]
Step 2: Swap 10 and 1 → [1, 5, 3, 4, | 10]
        Heapify(0) → [5, 4, 3, 1, | 10]
Step 3: Swap 5 and 1 → [1, 4, 3, | 5, 10]
        Heapify(0) → [4, 1, 3, | 5, 10]
Step 4: Swap 4 and 3 → [3, 1, | 4, 5, 10]
        Heapify(0) → [3, 1, | 4, 5, 10]
Step 5: Swap 3 and 1 → [1, | 3, 4, 5, 10]
Done: [1, 3, 4, 5, 10]
```

## Summary

### Key Concepts Learned:
1. **Heap Structure**: Array representation with parent/child formulas
2. **Max Heap Property**: Parent >= Children
3. **Heapify**: Fix violations by swapping down
4. **Build Heap**: Start from `n//2-1` (last internal node)
5. **HeapSort**: Build heap → Extract max repeatedly

### Time Complexity:
- Heapify: O(log n)
- Build heap: O(n)
- HeapSort: O(n log n)

### Space Complexity:
- O(1) - In-place sorting
- O(log n) if counting recursion stack

### When to Use HeapSort:
- Need O(n log n) worst case (better than QuickSort worst case)
- Need in-place sorting
- Finding kth largest/smallest elements
- Priority queue operations
