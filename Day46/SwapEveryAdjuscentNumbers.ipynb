24. Swap Nodes in Pairs

Given a linked list, swap every two adjacent nodes and return its head. You must solve the problem without modifying the values in the list's nodes (i.e., only nodes themselves may be changed.)

Example 1:

Input: head = [1,2,3,4]

Output: [2,1,4,3]

Explanation:



Example 2:

Input: head = []

Output: []

Example 3:

Input: head = [1]

Output: [1]

Example 4:

Input: head = [1,2,3]

Output: [2,1,3]

 

Constraints:

The number of nodes in the list is in the range [0, 100].
0 <= Node.val <= 100

## problem statement

for this solution we swap element with its next element and skip it and agin seap and skip one element.

- if only one element or empty we return the exact head.
if not head or not head.next:
    return head

dummy = ListNode(0)
dummy.next = head
prev = dummy

-we create a while loop with prev and prev.next.next
- swap prev.next and prev.next.next by
    first =  prev.next
    second = prev.next.next

    first.next = second.next
    second.next = first
    prev.next = second

    prev = first
- return dummy.next


```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        if not head or not head.next:
            return head
        
        dummy = ListNode(0)
        dummy.next = head
        prev = dummy

        while prev.next and prev.next.next:
            first = prev.next
            second = prev.next.next

            first.next = second.next
            second.next = first
            prev.next = second

            prev = first

        return dummy.next
    
# Test case 1: [1,2,3,4] -> [2,1,4,3]
# Helper function to print linked list values
def print_list(head):
    values = []
    current = head
    while current:
        values.append(current.val)
        current = current.next
    return values

node1 = ListNode(1)
node1.next = ListNode(2)
node1.next.next = ListNode(3)
node1.next.next.next = ListNode(4)

# Test case 2: [] -> []
node2 = None

# Test case 3: [1] -> [1]
node3 = ListNode(1)

# Test case 4: [1,2,3] -> [2,1,3]
node4 = ListNode(1)
node4.next = ListNode(2)
node4.next.next = ListNode(3)

s = Solution()
print(print_list(s.swapPairs(node1)))
print(print_list(s.swapPairs(node2)))
print(print_list(s.swapPairs(node3)))
print(print_list(s.swapPairs(node4)))
```
