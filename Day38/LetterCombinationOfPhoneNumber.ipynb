17. Letter Combinations of a Phone Number

Given a string containing digits from 2-9 inclusive, return all possible letter combinations that the number could represent. Return the answer in any order.

A mapping of digits to letters (just like on the telephone buttons) is given below. Note that 1 does not map to any letters.


 

Example 1:

Input: digits = "23"
Output: ["ad","ae","af","bd","be","bf","cd","ce","cf"]
Example 2:

Input: digits = "2"
Output: ["a","b","c"]
 

Constraints:

1 <= digits.length <= 4
digits[i] is a digit in the range ['2', '9'].

problem understanding is each number on phone has letter mapped with them. for this we use the dict to intialize them
nummap = {2:"abc",3:"def", ...}

for combinations we need do a recursion fun(digits,index) with base condition of indext reach the len of digits,
here for each recursion level, we iterate through the letters of digit and returns this list and resuse for the next recurion level

def recu(digits, index)

```python
def letterCombinations(digits):

    if digits=="":
        return []

    numMap = {"2":"abc", "3": "def", "4":"ghi", "5":"jkl","6":"mno", "7": "pqrs","8":"tuv", "9":"wxyz"}

    def combine(digits,index):
        res = []
        if index >= len(digits):
            return [""]

        nex_combo = combine(digits,index+1)

        alpha = numMap.get(digits[index])

        for a in alpha:
            for b in nex_combo:
                res.append(a+b)
        
        return res
        
    return combine(digits,0)

print(letterCombinations("23"))
```

```python
from typing import List

class Solution:
    digit_to_letter = {
        "2": ["a", "b", "c"],
        "3": ["d", "e", "f"],
        "4": ["g", "h", "i"],
        "5": ["j", "k", "l"],
        "6": ["m", "n", "o"],
        "7": ["p", "q", "r", "s"],
        "8": ["t", "u", "v"],
        "9": ["w", "x", "y", "z"],
    }
    def letterCombinations(self, digits: str) -> List[str]:
        combination_list = None
        for i in range(len(digits)):
            digit_i = digits[i]
            if combination_list is None:
                combination_list = self.digit_to_letter[digit_i]
            else:
                new_combination_list = [] 
                for c in combination_list:
                    for l in self.digit_to_letter[digit_i]:
                        new_combination_list.append(c+l)
                combination_list = new_combination_list
        return combination_list

s = Solution()
print(s.letterCombinations("23"))
```
