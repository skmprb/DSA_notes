Given the roots of two binary trees p and q, write a function to check if they are the same or not.

Two binary trees are considered the same if they are structurally identical, and the nodes have the same value.

 

Example 1:


Input: p = [1,2,3], q = [1,2,3]
Output: true
Example 2:


Input: p = [1,2], q = [1,null,2]
Output: false
Example 3:


Input: p = [1,2,1], q = [1,1,2]
Output: false
 

Constraints:

The number of nodes in both trees is in the range [0, 100].
-104 <= Node.val <= 104

```python
# Definition for a binary tree node.

from typing import Optional

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def isSameTree(self, p: Optional[TreeNode], q: Optional[TreeNode]) -> bool:
        
        if not p and not q:
            return True
        if not p or not q:
            return False
        return p.val == q.val and self.isSameTree(p.left, q.left) and self.isSameTree(p.right, q.right)
```

```python
# Test cases
sol = Solution()

# Example 1: [1,2,3] and [1,2,3]
p1 = TreeNode(1, TreeNode(2), TreeNode(3))
q1 = TreeNode(1, TreeNode(2), TreeNode(3))
print(f"Example 1: {sol.isSameTree(p1, q1)}")  # Expected: True

# Example 2: [1,2] and [1,null,2]
p2 = TreeNode(1, TreeNode(2), None)
q2 = TreeNode(1, None, TreeNode(2))
print(f"Example 2: {sol.isSameTree(p2, q2)}")  # Expected: False

# Example 3: [1,2,1] and [1,1,2]
p3 = TreeNode(1, TreeNode(2), TreeNode(1))
q3 = TreeNode(1, TreeNode(1), TreeNode(2))
print(f"Example 3: {sol.isSameTree(p3, q3)}")  # Expected: False
```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def isSameTree(self, p: Optional[TreeNode], q: Optional[TreeNode]) -> bool:
        
        def dfs(p, q):
            if not p and not q:
                return True

            if not (p and q):
                return False

            if p.val != q.val:
                return False

            return dfs(p.left, q.left) and dfs(p.right, q.right)

        return dfs(p, q)
```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def isSameTree(self, p: Optional[TreeNode], q: Optional[TreeNode]) -> bool:
        stack_p = [p]
        stack_q = [q]

        while stack_p and stack_q:
            pop_p = stack_p.pop()
            pop_q = stack_q.pop()

            if not pop_p and not pop_q:  # both None → fine, skip
                continue
            if not pop_p or not pop_q:   # one is None, other isn't → different
                return False
            if pop_p.val != pop_q.val:   # both exist but values differ
                return False

            # always append both children, even if None
            stack_p.append(pop_p.left)
            stack_p.append(pop_p.right)
            stack_q.append(pop_q.left)
            stack_q.append(pop_q.right)

        return True
```
