415. Add Strings

Given two non-negative integers, num1 and num2 represented as string, return the sum of num1 and num2 as a string.

You must solve the problem without using any built-in library for handling large integers (such as BigInteger). You must also not convert the inputs to integers directly.

 

Example 1:

Input: num1 = "11", num2 = "123"
Output: "134"
Example 2:

Input: num1 = "456", num2 = "77"
Output: "533"
Example 3:

Input: num1 = "0", num2 = "0"
Output: "0"
 

Constraints:

1 <= num1.length, num2.length <= 104
num1 and num2 consist of only digits.
num1 and num2 don't have any leading zeros except for the zero itself.

```python
class Solution:
    def addStrings(self, num1: str, num2: str) -> str:
        # Initialize pointers to the last character of each string (rightmost digits)
        i, j = len(num1) - 1, len(num2) - 1
        # List to store the result digits
        ans = []
        # Carry variable for addition
        c = 0
        # Continue while there are digits to process or there's a carry
        while i >= 0 or j >= 0 or c:
            # Get digit from num1 or 0 if we've processed all digits
            a = 0 if i < 0 else int(num1[i])
            # Get digit from num2 or 0 if we've processed all digits
            b = 0 if j < 0 else int(num2[j])
            # Calculate sum with carry, divmod returns (quotient, remainder)
            c, v = divmod(a + b + c, 10)
            # Append the digit (remainder) to result
            ans.append(str(v))
            # Move to next digits (going left)
            i, j = i - 1, j - 1
        # Reverse the result since we built it backwards and join into string
        return "".join(ans[::-1])

    def subStrings(self, num1: str, num2: str) -> str:
        # Get lengths of both strings
        m, n = len(num1), len(num2)
        # Check if result will be negative (num1 < num2)
        neg = m < n or (m == n and num1 < num2)
        # If negative, swap numbers to ensure we subtract smaller from larger
        if neg:
            num1, num2 = num2, num1
        # Initialize pointers to rightmost digits
        i, j = len(num1) - 1, len(num2) - 1
        # List to store result digits
        ans = []
        # Borrow variable for subtraction
        c = 0
        # Process all digits of the larger number
        while i >= 0:
            # Calculate difference: current digit - borrow - corresponding digit from num2
            c = int(num1[i]) - c - (0 if j < 0 else int(num2[j]))
            # If negative, add 10 and set borrow for next iteration
            ans.append(str((c + 10) % 10))
            # Set borrow flag: 1 if we needed to borrow, 0 otherwise
            c = 1 if c < 0 else 0
            # Move to next digits (going left)
            i, j = i - 1, j - 1
        # Remove leading zeros from result (except if result is just "0")
        while len(ans) > 1 and ans[-1] == '0':
            ans.pop()
        # Add negative sign if original result should be negative
        if neg:
            ans.append('-')
        # Reverse and join the result
        return ''.join(ans[::-1])

# Create instance of Solution class
s = Solution()
# Test the addStrings method with two number strings
print(s.addStrings("121322","131323"))
```
