231. Power of Two

Given an integer n, return true if it is a power of two. Otherwise, return false.

An integer n is a power of two, if there exists an integer x such that n == 2x.

 

Example 1:

Input: n = 1
Output: true
Explanation: 20 = 1
Example 2:

Input: n = 16
Output: true
Explanation: 24 = 16
Example 3:

Input: n = 3
Output: false
 

Constraints:

-231 <= n <= 231 - 1
 

Follow up: Could you solve it without loops/recursion?

```python
# Function to check if a number is a power of 2
def ispowerofTwo(n):
    # First check if the number is positive (powers of 2 are always positive)
    if n > 0:
        # Use bitwise AND operation: n & (n-1)
        # For powers of 2, this operation always equals 0
        # Example: 8 (1000) & 7 (0111) = 0000 = 0
        # Example: 6 (0110) & 5 (0101) = 0100 = 4 (not 0, so not power of 2)
        if (n&(n-1)) == 0:
            return True
        else:
            return False
    else:
        # Return False for zero, negative numbers, or any non-positive input
        return False

# Test the function with 16 (which is 2^4, so should return True)
print(ispowerofTwo(16))
```

```python
# The bitwise trick (n & (n-1)) == 0 only works for powers of 2
# For powers of 3 or 4, you need different approaches:
# - Powers of 3: Use logarithms or repeated division
# - Powers of 4: Can use bitwise but different logic (n > 0 and (n & (n-1)) == 0 and (n & 0x55555555) != 0)
def ispowerofTwo(n):
    return n > 0 and (n&(n-1)) == 0

print(ispowerofTwo(16))
```
