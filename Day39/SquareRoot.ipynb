Given a non-negative integer x, return the square root of x rounded down to the nearest integer. The returned integer should be non-negative as well.

You must not use any built-in exponent function or operator.

For example, do not use pow(x, 0.5) in c++ or x ** 0.5 in python.
 

Example 1:

Input: x = 4
Output: 2
Explanation: The square root of 4 is 2, so we return 2.
Example 2:

Input: x = 8
Output: 2
Explanation: The square root of 8 is 2.82842..., and since we round it down to the nearest integer, 2 is returned.

### Solution 1: Binary Search (Your Approach - Fixed)
Time: O(log n) | Space: O(1)

```python
def mySqrt(x):
    # Time Complexity: O(log n) - Binary search reduces search space by half each iteration
    # Space Complexity: O(1) - Only using constant extra space for variables
    if x < 2:
        return x
    
    left, right = 1, x // 2
    ans = 0
    
    while left <= right:
        mid = (left + right) // 2  # Integer division
        
        if mid * mid == x:
            return mid
        elif mid * mid < x:
            left = mid + 1
            ans = mid  # Save last valid answer
        else:
            right = mid - 1
    
    return ans

# Test
print(mySqrt(8))   # 2
print(mySqrt(4))   # 2
print(mySqrt(25))  # 5
print(mySqrt(1))   # 1
```

### Solution 2: Optimized Binary Search (Avoid Overflow)
Time: O(log n) | Space: O(1)

Instead of mid*mid, use mid == x//mid to avoid overflow for large numbers

```python
def mySqrt_optimized(x):
    if x < 2:
        return x
    
    left, right = 1, x // 2
    
    while left <= right:
        mid = (left + right) // 2
        
        if mid == x // mid:  # Avoids overflow
            return mid
        elif mid < x // mid:
            left = mid + 1
        else:
            right = mid - 1
    
    return right  # right will be the answer

# Test
print(mySqrt_optimized(8))   # 2
print(mySqrt_optimized(4))   # 2
print(mySqrt_optimized(25))  # 5

# Time Complexity: O(log n) - Binary search reduces search space by half in each iteration
# Space Complexity: O(1) - Only uses constant extra space for variables (left, right, mid)
# Program size: Approximately 15-20 lines of code, roughly 400-500 bytes in memory
```

### Solution 3: Newton's Method (Fastest!)
Time: O(log n) but faster in practice | Space: O(1)

Uses Newton-Raphson method: x_new = (x_old + n/x_old) / 2

```python
def mySqrt_newton(x):
    if x < 2:
        return x
    
    # Start with x/2 as initial guess
    guess = x
    
    while guess * guess > x:
        guess = (guess + x // guess) // 2
    
    return guess

# Test
print(mySqrt_newton(8))   # 2
print(mySqrt_newton(4))   # 2
print(mySqrt_newton(25))  # 5

# Time complexity O(log log n) : Newton's method has quadratic convergence
# Space complexity O(1) : constant extra space used
```

### Comparison:

3. **Newton's Method**: Fastest convergence, but less intuitive
