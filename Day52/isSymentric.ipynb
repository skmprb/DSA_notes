101. Symmetric Tree

Given the root of a binary tree, check whether it is a mirror of itself (i.e., symmetric around its center).

 

Example 1:

- Input: root = [1,2,2,3,4,4,3]
- Output: true

Example 2:

- Input: root = [1,2,2,null,3,null,3]
- Output: false
 

Constraints:

The number of nodes in the tree is in the range [1, 1000].
-100 <= Node.val <= 100
 

Follow up: Could you solve it both recursively and iteratively?

## Problem Understanding and logic

I have a root and nodes, here we have check whether its is symmentric like we have check as:
- if root.left == root.right , in next iteration
- if root.left.left == root.right.right and root.left.right == root.right.left
- and contintue

but here as depth grows the nodes also grow a lot, how would we keep the track of these?

we need no need to track all, but we have to check the left node and right node of their lefts and rights are symentric or not.

we can conquour this using helper function. by recursivly checking the symentry

the base case for this is:

so here we have check three cases as:

- if both left and right tree are none -> true
- if even any one have one sub tree and other don't - > false
- if both have sub tree and their values are not equal -> false



```python
from typing import Optional

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def isSymmetric(self, root: Optional[TreeNode]) -> bool:

        if root is None:
            return True

        def isMirror(lefttree, righttree):
            if lefttree is None and righttree is None:
                return True
            
            if lefttree is None or righttree is None or lefttree.val != righttree.val:
                return False
            
            return isMirror(lefttree.left, righttree.right) and isMirror(lefttree.right, righttree.left )

        return isMirror(root.left, root.right)


if __name__ == "__main__":
    
    root = TreeNode(10)
    root.left = TreeNode(5)
    root.right = TreeNode(5)
    root.left.left = TreeNode(2)
    root.right.right = TreeNode(2)

    s = Solution()
    print("true" if s.isSymmetric(root) else "false")
        
```

```python
from typing import Optional
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def isSymmetric(self, root: Optional[TreeNode]) -> bool:

        if root is None:
            return True

        q = deque()
        q.append((root.left, root.right))
        while q:
            leftnode, rightnode = q.popleft()

            if leftnode is None and rightnode is None:
                continue

            if leftnode is None or rightnode is None or leftnode.val != rightnode.val:
                return False

            q.append((leftnode.left, rightnode.right))
            q.append((leftnode.right, rightnode.left))
        
        return True


if __name__ == "__main__":
    
    root = TreeNode(10)
    root.left = TreeNode(5)
    root.right = TreeNode(5)
    root.left.left = TreeNode(2)
    root.right.right = TreeNode(2)

    s = Solution()
    print("true" if s.isSymmetric(root) else "false")
        
```
