# Priority Queue with Heap - Complete Guide

## What is a Priority Queue?
A **Priority Queue** is a data structure where each element has a priority, and elements are served based on their priority (not insertion order).

**Key Operations:**
- `insert(element, priority)` - Add element with priority
- `extract_max()` or `extract_min()` - Remove and return highest/lowest priority element
- `peek()` - View highest/lowest priority element without removing

**Real-world Examples:**
- Hospital emergency room (critical patients first)
- CPU task scheduling (high priority tasks first)
- Dijkstra's shortest path algorithm
- Huffman coding

---

## Why Use Heap for Priority Queue?

| Operation | Array (unsorted) | Array (sorted) | Heap |
|-----------|------------------|----------------|------|
| Insert | O(1) | O(n) | **O(log n)** |
| Extract Max/Min | O(n) | O(1) | **O(log n)** |
| Peek | O(n) | O(1) | **O(1)** |

**Heap provides the best balance!**

## Part 2: Insert Operation (Heapify Up)

### Concept: Inserting into Priority Queue
1. Add new element at the **end** of array
2. **Bubble up** (heapify up) to restore heap property
3. Compare with parent, swap if needed, repeat

```
Insert 40 into [50, 30, 20, 15, 10]:

Step 1: Add at end
       50
      /  \
    30    20
   /  \   /
  15  10 40  ← New element

Step 2: Compare with parent (20)
40 > 20, so swap
       50
      /  \
    30    40  ← Swapped
   /  \   /
  15  10 20

Step 3: Compare with parent (50)
40 < 50, stop
```

**Key Formula:** Parent of index `i` = `(i-1) // 2`

### Exercise 2.1: Heapify Up (Bubble Up)
Move element up the heap until heap property is satisfied.

````python
def heapify_up(heap, index):
    # TODO: Bubble up element at 'index' to correct position
    # While index > 0:
    #   Calculate parent_index = (index - 1) // 2
    #   If heap[index] > heap[parent_index]:
    #       Swap them
    #       Update index = parent_index
    #   Else:
    #       Break
    while index > 0:
        j = (index - 1) // 2
        if heap[index] > heap[j]:
            heap[index], heap[j] = heap[j], heap[index]
            index = j
        else:
            break

# Test
heap = [50, 30, 20, 15, 10, 40]  # 40 is at wrong position
print("Before:", heap)
heapify_up(heap, 5)  # Fix position of 40
print("After:", heap)  # Should be [50, 30, 40, 15, 10, 20]
````

## Part 3: Extract Max Operation (Heapify Down)

### Concept: Removing Highest Priority Element
1. Save root element (max)
2. Move **last element** to root
3. Remove last element
4. **Heapify down** from root to restore heap property

```
Extract max from [50, 30, 20, 15, 10]:

Step 1: Save 50, move 10 to root
       10
      /  \
    30    20
   /
  15

Step 2: Heapify down - 10 < 30, swap
       30
      /  \
    10    20
   /
  15

Step 3: Continue - 10 < 15, swap
       30
      /  \
    15    20
   /
  10
```

### Exercise 3.1: Heapify Down
Move element down the heap to correct position.

````python
def heapify_down(heap, index):
    # TODO: Heapify down from 'index'
    # n = len(heap)
    # largest = index
    # left = 2 * index + 1
    # right = 2 * index + 2
    # Find largest among parent and children
    # If largest != index, swap and recursively heapify_down
    
    n = len(heap)
    largest = index
    left = (2 * index)+1
    right = (2 * index)+2

    if left < n and heap[left] > heap[largest]:
        largest = left
    
    if right < n and heap[right] > heap[largest]:
        largest = right
    
    if largest != index:
        heap[largest] , heap[index] = heap[index], heap[largest]
        return heapify_down(heap, largest)


# Test
heap = [10, 30, 20, 15]
print("Before:", heap)
heapify_down(heap, 0)
print("After:", heap)  # Should be [30, 15, 20, 10]
````

## Summary

### Key Concepts:

**1. Priority Queue Operations:**
- Insert: O(log n) - Add and heapify up
- Extract: O(log n) - Remove root and heapify down
- Peek: O(1) - Just return root

**2. Max vs Min Heap:**
- Max Heap: Parent >= Children (largest at root)
- Min Heap: Parent <= Children (smallest at root)

**3. Heapify Up vs Down:**
- Heapify Up: Used in insert (bubble up from bottom)
- Heapify Down: Used in extract (bubble down from top)

**4. Python heapq:**
- Always min heap
- For max heap: negate values
- heapify() is O(n), faster than n inserts

**5. Custom Priority:**
- Use tuples: (priority, value)
- Or custom class with __lt__ method

### Common Patterns:

**Pattern 1: Top K Largest**
```python
heap = nums[:k]
heapq.heapify(heap)  # Min heap of size k
for num in nums[k:]:
    if num > heap[0]:
        heapq.heapreplace(heap, num)
```

**Pattern 2: Merge K Sorted Lists**
```python
heap = [(lists[i][0], i, 0) for i in range(k) if lists[i]]
heapq.heapify(heap)
while heap:
    val, list_idx, elem_idx = heapq.heappop(heap)
    # Process val
    if elem_idx + 1 < len(lists[list_idx]):
        heapq.heappush(heap, (lists[list_idx][elem_idx+1], list_idx, elem_idx+1))
```

**Pattern 3: Two Heaps (Median)**
```python
small = []  # Max heap (negate values)
large = []  # Min heap
# Keep sizes balanced
# Median is at heap tops
```

### When to Use What:

| Need | Use | Why |
|------|-----|-----|
| Top k largest | Min heap size k | Keep k largest, remove smallest |
| Top k smallest | Max heap size k | Keep k smallest, remove largest |
| Kth largest | Min heap size k | kth largest is at root |
| Median | Two heaps | Split data in half |
| Merge k sorted | Min heap | Always pick smallest available |

### Max Heap vs Min Heap Decision:

**Use Max Heap when:**
- Need largest element repeatedly
- Scheduling (highest priority first)
- Top k smallest (keep k smallest, remove largest)

**Use Min Heap when:**
- Need smallest element repeatedly
- Dijkstra's algorithm (shortest distance)
- Top k largest (keep k largest, remove smallest)

### Time Complexity Comparison:

| Method | Build Heap | Insert n times |
|--------|------------|----------------|
| heapify() | O(n) | - |
| n × insert | - | O(n log n) |

**Always use heapify() when converting existing list!**
