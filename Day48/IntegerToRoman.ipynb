Seven different symbols represent Roman numerals with the following values:

```
Symbol	Value
I	1
V	5
X	10
L	50
C	100
D	500
M	1000
```

Roman numerals are formed by appending the conversions of decimal place values from highest to lowest. Converting a decimal place value into a Roman numeral has the following rules:

- If the value does not start with 4 or 9, select the symbol of the maximal value that can be subtracted from the input, append that symbol to the result, subtract its value, and convert the remainder to a Roman numeral.
- If the value starts with 4 or 9 use the subtractive form representing one symbol subtracted from the following symbol, for example, 4 is 1 (I) less than 5 (V): IV and 9 is 1 (I) less than 10 (X): IX. Only the following subtractive forms are used: 4 (IV), 9 (IX), 40 (XL), 90 (XC), 400 (CD) and 900 (CM).
- Only powers of 10 (I, X, C, M) can be appended consecutively at most 3 times to represent multiples of 10. You cannot append 5 (V), 50 (L), or 500 (D) multiple times. If you need to append a symbol 4 times use the subtractive form.

# Learning Lab: Integer to Roman Conversion

## Understanding the Problem

Roman numerals use 7 basic symbols:
- I = 1, V = 5, X = 10, L = 50, C = 100, D = 500, M = 1000

**Special Rule - Subtractive Notation:**
When a smaller value appears before a larger value, we SUBTRACT it:
- IV = 4 (5 - 1)
- IX = 9 (10 - 1)
- XL = 40 (50 - 10)
- XC = 90 (100 - 10)
- CD = 400 (500 - 100)
- CM = 900 (1000 - 100)

## Why Greedy Algorithm Works

**Key Insight:** Roman numerals are written from largest to smallest (left to right).

**Strategy:** Always use the LARGEST possible value first, then move to smaller values.

### Example: Converting 3749

Think of it like making change with coins, but using Roman "coins":

```
Available "coins": 1000, 900, 500, 400, 100, 90, 50, 40, 10, 9, 5, 4, 1
Symbols:           M,   CM,  D,   CD,  C,   XC, L,  XL, X,  IX, V, IV, I

Number: 3749

Step 1: Can I use 1000? Yes, 3 times!
  3749 - 3000 = 749
  Result: "MMM"

Step 2: Can I use 900? No (749 < 900)

Step 3: Can I use 500? Yes, 1 time!
  749 - 500 = 249
  Result: "MMMD"

Step 4: Can I use 400? No (249 < 400)

Step 5: Can I use 100? Yes, 2 times!
  249 - 200 = 49
  Result: "MMMDCC"

Step 6: Can I use 90? No (49 < 90)
Step 7: Can I use 50? No (49 < 50)

Step 8: Can I use 40? Yes, 1 time!
  49 - 40 = 9
  Result: "MMMDCCXL"

Step 9: Can I use 10? No (9 < 10)

Step 10: Can I use 9? Yes, 1 time!
  9 - 9 = 0
  Result: "MMMDCCXLIX"

Done! ✓
```

## Why Include Subtractive Forms in the Mapping?

**Without subtractive forms:**
- 4 would be "IIII" (wrong!)
- 9 would be "VIIII" (wrong!)
- 40 would be "XXXX" (wrong!)

**With subtractive forms in mapping:**
- 4 → "IV" (correct!)
- 9 → "IX" (correct!)
- 40 → "XL" (correct!)

By including these special cases (4, 9, 40, 90, 400, 900) in our mapping,
the greedy algorithm automatically handles them correctly!

## Algorithm Derivation

### Step 1: Identify all possible values
Basic symbols: 1, 5, 10, 50, 100, 500, 1000
Subtractive forms: 4, 9, 40, 90, 400, 900

### Step 2: Sort in descending order
1000, 900, 500, 400, 100, 90, 50, 40, 10, 9, 5, 4, 1

### Step 3: Map each value to its symbol
M, CM, D, CD, C, XC, L, XL, X, IX, V, IV, I

### Step 4: Apply greedy algorithm
For each value (largest to smallest):
  - Calculate how many times it fits: count = num // value
  - Append the symbol that many times
  - Subtract from num: num -= value * count
  - Continue until num becomes 0

## Why This is Optimal

1. **Greedy choice is always correct** - Roman numerals are designed to be written largest to smallest
2. **No backtracking needed** - Once we use a value, we never need to reconsider
3. **Handles all edge cases** - Subtractive forms are built into the mapping
4. **Time complexity: O(1)** - Fixed 13 iterations maximum
5. **Space complexity: O(1)** - Fixed size arrays


## Implementation with Detailed Comments

````python
def intToRoman(num: int) -> str:
    """
    Convert an integer to Roman numeral using greedy algorithm.
    
    Time Complexity: O(1) - Fixed 13 iterations
    Space Complexity: O(1) - Fixed size arrays
    """
    
    # All possible values in descending order (including subtractive forms)
    values = [1000, 900, 500, 400, 100, 90, 50, 40, 10, 9, 5, 4, 1]
    
    # Corresponding Roman numeral symbols
    symbols = ["M", "CM", "D", "CD", "C", "XC", "L", "XL", "X", "IX", "V", "IV", "I"]
    
    result = ""
    
    # Greedy approach: use largest values first
    for i in range(len(values)):
        # How many times can we use this value?
        count = num // values[i]
        
        if count > 0:
            # Append the symbol 'count' times
            result += symbols[i] * count
            
            # Subtract the value we just used
            num -= values[i] * count
    
    return result

# Test cases
print(f"3749 → {intToRoman(3749)}")  # Expected: MMMDCCXLIX
print(f"58 → {intToRoman(58)}")      # Expected: LVIII
print(f"1994 → {intToRoman(1994)}")  # Expected: MCMXCIV
````

## Step-by-Step Trace for 1994

Let's trace through the algorithm with num = 1994:

````python
def intToRomanWithTrace(num: int) -> str:
    """Detailed trace of the conversion process"""
    
    values = [1000, 900, 500, 400, 100, 90, 50, 40, 10, 9, 5, 4, 1]
    symbols = ["M", "CM", "D", "CD", "C", "XC", "L", "XL", "X", "IX", "V", "IV", "I"]
    
    result = ""
    original_num = num
    
    print(f"Converting {num} to Roman numeral:\n")
    
    for i in range(len(values)):
        count = num // values[i]
        
        if count > 0:
            print(f"Step: num={num}, value={values[i]}, symbol={symbols[i]}")
            print(f"  → {num} // {values[i]} = {count} (use {symbols[i]} {count} time(s))")
            
            result += symbols[i] * count
            num -= values[i] * count
            
            print(f"  → Result so far: \"{result}\"")
            print(f"  → Remaining: {num}\n")
    
    print(f"Final Result: {original_num} → \"{result}\"")
    return result

# Trace example
intToRomanWithTrace(1994)
````

## Alternative Implementation: Using Tuples

````python
def intToRomanAlt(num: int) -> str:
    """Alternative implementation using list of tuples"""
    
    # Combine values and symbols into tuples
    mapping = [
        (1000, "M"), (900, "CM"), (500, "D"), (400, "CD"),
        (100, "C"), (90, "XC"), (50, "L"), (40, "XL"),
        (10, "X"), (9, "IX"), (5, "V"), (4, "IV"), (1, "I")
    ]
    
    result = ""
    
    for value, symbol in mapping:
        count = num // value
        if count:
            result += symbol * count
            num -= value * count
    
    return result

# Test
print(intToRomanAlt(3749))  # MMMDCCXLIX
````

## Key Takeaways

1. **Greedy algorithms work when the greedy choice is always optimal**
   - Roman numerals are designed to be written largest to smallest
   - This makes greedy the perfect approach

2. **Include special cases in your data structure**
   - By adding subtractive forms (4, 9, 40, 90, 400, 900) to the mapping
   - We eliminate the need for special case handling in code

3. **Order matters**
   - Always process from largest to smallest
   - This ensures we use the maximum value at each step

4. **Time complexity is constant**
   - Even though we have a loop, it's always 13 iterations max
   - This makes it O(1), not O(n)

5. **This pattern applies to other problems**
   - Coin change (making change with minimum coins)
   - Activity selection
   - Huffman coding
